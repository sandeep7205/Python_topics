```python
# Question 1: Write a function to calculate the factorial of a number.
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)

# Example usage:
print(factorial(5))  # Expected output: 120

# Question 2: Given a list of numbers, write a function to find the largest number.
def find_largest(numbers):
    max_number = numbers[0]
    for number in numbers[1:]:
        if number > max_number:
            max_number = number
    return max_number

# Example usage:
print(find_largest([1, 5, 2, 8, 3]))  # Expected output: 8

# Question 3: Write a function that checks if a given word is a palindrome.
def is_palindrome(word):
    return word == word[::-1]

# Example usage:
print(is_palindrome("radar"))  # Expected output: True
print(is_palindrome("python"))  # Expected output: False

# Question 4: Write a function to merge two sorted lists into a single sorted list.
def merge_sorted_lists(list1, list2):
    i = j = 0
    merged_list = []
    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:
            merged_list.append(list1[i])
            i += 1
        else:
            merged_list.append(list2[j])
            j += 1
    merged_list.extend(list1[i:])
    merged_list.extend(list2[j:])
    return merged_list

# Example usage:
print(merge_sorted_lists([1, 3, 5], [2, 4, 6]))  # Expected output: [1, 2, 3, 4, 5, 6]

# Question 5: Write a Python function that takes a number as a parameter and checks if the number is prime or not.
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Example usage:
print(is_prime(29))  # Expected output: True
print(is_prime(12))  # Expected output: False

# Question 6: Write a function to convert a temperature from Celsius to Fahrenheit.
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

# Example usage:
print(celsius_to_fahrenheit(0))  # Expected output: 32.0

# Question 7: Write a function that returns the number of vowels found in a string.
def count_vowels(s):
    count = 0
    for char in s.lower():
        if char in 'aeiou':
            count += 1
    return count

# Example usage:
print(count_vowels("Hello World"))  # Expected output: 3

# Question 8: Write a function to reverse a string.
def reverse_string(s):
    return s[::-1]

# Example usage:
print(reverse_string("hello"))  # Expected output: "olleh"

# Question 9: Write a function that accepts a string and calculates the number of upper case letters and lower case letters.
def case_counter(s):
    upper, lower = 0, 0
    for char in s:
        if char.isupper():
            upper += 1
        elif char.islower():
            lower += 1
    return upper, lower

# Example usage:
print(case_counter("Hello World"))  # Expected output: (2, 8)

# Question 10: Write a function that takes a list and returns a new list with unique elements of the first list.
def unique_list(lst):
    return list(set(lst))

# Example usage:
print(unique_list([1, 2, 2, 3, 3, 3, 4]))  # Expected output: [1, 2, 3, 4]
```
