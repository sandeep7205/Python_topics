```python
# Question 1: Write a function to reverse a string in Python.
def reverse_string(s):
    return s[::-1]

# Question 2: Write a Python function to check if a number is prime.
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Question 3: Write a Python function that takes a list and returns a new list with unique elements of the first list.
def unique_list(l):
    return list(set(l))

# Question 4: Write a Python program to find the Fibonacci sequence up to n.
def fibonacci(n):
    fib_sequence = [0, 1]
    while len(fib_sequence) < n:
        fib_sequence.append(fib_sequence[-1] + fib_sequence[-2])
    return fib_sequence[:n]

# Question 5: Write a Python function that checks whether a passed string is palindrome or not.
def is_palindrome(s):
    return s == s[::-1]

# Question 6: Write a Python function to sum all the numbers in a list.
def sum_list(items):
    return sum(items)

# Question 7: Write a Python function to multiply all the numbers in a list.
def multiply_list(items):
    result = 1
    for item in items:
        result *= item
    return result

# Question 8: Write a Python function that accepts a string and calculate the number of upper case letters and lower case letters.
def string_test(s):
    d = {"UPPER_CASE": 0, "LOWER_CASE": 0}
    for c in s:
        if c.isupper():
            d["UPPER_CASE"] += 1
        elif c.islower():
            d["LOWER_CASE"] += 1
    return d

# Question 9: Write a Python program to sort a list of tuples by the second item.
def sort_tuples(list_of_tuples):
    return sorted(list_of_tuples, key=lambda x: x[1])

# Question 10: Write a Python function to check whether a string is an anagram of another string.
def is_anagram(str1, str2):
    return sorted(str1) == sorted(str2)
```
