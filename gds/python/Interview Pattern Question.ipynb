```python
# Question 1: Print a square pattern with asterisks (*).
# This function prints a square of 'n' size using asterisks. 
# It iterates over a range of 'n' and prints 'n' asterisks followed by a space in each row.
def print_square(n):
    for i in range(n):
        print('* ' * n)  # Print 'n' asterisks followed by a space, repeated 'n' times.

# Example usage:
# print_square(4)

# Question 2: Print a right-angled triangle pattern with asterisks (*).
# This function prints a right-angled triangle pattern.
# For each row 'i', it prints 'i' asterisks, thereby increasing the number of asterisks in each subsequent row.
def print_right_angled_triangle(n):
    for i in range(1, n + 1):
        print('* ' * i)  # Print 'i' asterisks in each row, where 'i' increases from 1 to 'n'.

# Example usage:
# print_right_angled_triangle(5)

# Question 3: Print an inverted right-angled triangle pattern with asterisks (*).
# This function prints an inverted right-angled triangle.
# It starts with 'n' asterisks and decreases the count by one in each subsequent row.
def print_inverted_triangle(n):
    for i in range(n, 0, -1):
        print('* ' * i)  # Print 'i' asterisks in each row, where 'i' decreases from 'n' to 1.

# Example usage:
# print_inverted_triangle(5)

# Question 4: Print a pyramid pattern with asterisks (*).
# This function prints a pyramid pattern centered on the screen.
# For each row 'i', it prints 'n-i' spaces followed by 'i' asterisks, thus forming a pyramid shape.
def print_pyramid(n):
    for i in range(1, n + 1):
        print(' ' * (n - i) + '* ' * i)  # Print 'n-i' spaces followed by 'i' asterisks, forming a pyramid.

# Example usage:
# print_pyramid(5)

# Question 5: Print a diamond pattern with asterisks (*).
# This function prints a diamond pattern by first increasing the number of asterisks to create the upper half,
# and then decreasing the number of asterisks to create the lower half of the diamond.
def print_diamond(n):
    # Upper half of the diamond
    for i in range(1, n + 1):
        print(' ' * (n - i) + '* ' * i)  # Increase asterisks to form the upper half.
    # Lower half of the diamond
    for i in range(n - 1, 0, -1):
        print(' ' * (n - i) + '* ' * i)  #
```
