
# Solutions for Module 11.1 Error and Exception Handling: Coding Questions

This notebook contains solutions to coding questions related to error and exception handling in Python.


```

def safe_division(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero."
    
```

```

def access_list_element(lst, index):
    try:
        return lst[index]
    except IndexError:
        return "Index out of bounds."
    
```

```

def parse_string_to_int(s):
    try:
        return int(s)
    except ValueError:
        return "Could not convert string to integer."
    
```
