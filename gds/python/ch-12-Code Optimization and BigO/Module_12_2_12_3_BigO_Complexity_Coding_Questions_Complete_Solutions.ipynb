
# Complete Solutions for Module 12.2 & 12.3 BigO Notations and Complexity Rules: Coding Questions

This notebook provides complete solutions and complexity analysis for coding questions related to BigO notations and complexity rules.


```

# Question 1: Check if an item is in a list
def item_in_list(lst, item):
    for i in lst:
        if i == item:
            return True
    return False
# Time Complexity: O(n), where n is the number of items in the list.
```

```

# Question 2: Find the largest number in a list
def find_largest(lst):
    largest = lst[0]
    for num in lst[1:]:
        if num > largest:
            largest = num
    return largest
# Time Complexity: O(n), where n is the number of items in the list.
```

```

# Question 3: Merge two sorted lists
def merge_sorted_lists(lst1, lst2):
    merged, i, j = [], 0, 0
    while i < len(lst1) and j < len(lst2):
        if lst1[i] < lst2[j]:
            merged.append(lst1[i])
            i += 1
        else:
            merged.append(lst2[j])
            j += 1
    merged.extend(lst1[i:])
    merged.extend(lst2[j:])
    return merged
# Time Complexity: O(n+m), Space Complexity: O(n+m), where n and m are the lengths of the lists.
```

```

# Question 4: Space complexity of calculating nth Fibonacci number recursively
def fib(n):
    if n <= 1:
        return n
    else:
        return fib(n-1) + fib(n-2)
# Space Complexity: O(n) due to the depth of the recursion stack.
```

```

# Question 5: Remove duplicates from a list
def remove_duplicates(lst):
    return list(set(lst))
# Time Complexity: O(n), where n is the number of items in the list.
```

```

# Question 6: Binary search algorithm
def binary_search(sorted_lst, item):
    left, right = 0, len(sorted_lst) - 1
    while left <= right:
        mid = (left + right) // 2
        if sorted_lst[mid] == item:
            return mid
        elif sorted_lst[mid] < item:
            left = mid + 1
        else:
            right = mid - 1
    return -1
# Time Complexity: O(log n), where n is the number of items in the list.
```

```

# Question 7: Bubble sort and its complexities
def bubble_sort(lst):
    n = len(lst)
    for i in range(n):
        for j in range(0, n-i-1):
            if lst[j] > lst[j+1]:
                lst[j], lst[j+1] = lst[j+1], lst[j]
    return lst
# Average and Worst-case Time Complexity: O(n^2), where n is the number of items in the list.
```

```

# Question 8: Calculate the sum of all numbers in a list
def sum_numbers(lst):
    return sum(lst)
# Time Complexity: O(n), Space Complexity: O(1), where n is the number of items in the list.
```
