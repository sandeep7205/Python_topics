```python
import numpy as np
```

```python
#2d array
two_d_arr = np.array([[3,4, 5],[1,2, 0]])
print(two_d_arr)
print(two_d_arr.shape)
```

```python
#3d array
three_d_arr = np.array([[[3,4, 5],[1,'2', 0]], [[36,44, 53],[21,52.89, 70]]])
print(three_d_arr)
print(three_d_arr.shape)
```

```python
# shape : shape of an array
print("shape of the array = ", two_d_arr.shape)
print("shape of the array = ", three_d_arr.shape)
```

```python
# ndim : dimensions of an array
print("Dimenions of the array = ", two_d_arr.ndim)
print("Dimenions of the array = ", three_d_arr.ndim)
```

```python
#size: total number of elements in the array
print("total number of elements in the array = ", two_d_arr.size)
print("total number of elements in the array = ", three_d_arr.size)
```

```python
#dtype: Data type of the array
print("Data type of the array = ", two_d_arr.dtype)
print("Data type of the array = ", three_d_arr.dtype)
```

```python
#astype: changes data type of the array
three_d_arr_n = three_d_arr.astype('float64')
print("changes data type of the array = ", three_d_arr_n.dtype)
```
