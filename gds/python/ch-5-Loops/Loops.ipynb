# Loops in Python

## Introduction

In Python, loops are a fundamental construct that allow you to repeat a block of code multiple times. This repetition is essential for tasks that require iterative processing, such as traversing data structures, performing calculations on multiple items, or automating repetitive tasks. Python offers two primary types of loops: `for` loops and `while` loops.

## `for` Loops

The `for` loop in Python is used to iterate over a sequence, which could be a list, tuple, dictionary, set, or string. This type of loop executes a set of statements once for each item in the sequence.

### Key Features:
- **Iterating through sequences**: Automatically handles the iteration over items in a sequence.
- **Range function**: Often used with the `range()` function for a specified number of iterations.
- **Loop control statements**: Includes `break`, `continue`, and `else` for additional control over the loop.


```python
# Looping from 0 to 4
for i in range(10):
    print(i)
```

```python
# Looping from 10 to 14
for i in range(10, 15):
    print(i)
```

```python
# Looping from 1 to 9, with a step of 2
for i in range(1, 10, 2):
    print(i)
```

```python
# Iterating through each character in a string
for char in "Hello":
    print(char)
```

```python
# Iterating through a string with index and character
for index, char in enumerate("World"):
    print(index, char)
```

```python
#check for even and odd


# Loop through numbers from 1 to 10
for num in range(1, 11):
    print(num)
    # Check if the number is even
    if num % 2 == 0:
        print(f"{num} is even")
    else:
        # Execute this block if the number is odd
        print(f"{num} is odd")
```

```python
#sum of even numbers

# Initialize a variable to hold the sum

total_sum = 0

# Loop to add even numbers from 1 to 10

for num in range(1, 11):
    # Add number to sum if it's even
    if num % 2 == 0:
        
        total_sum += num
print("Sum of even numbers from 1 to 10:", total_sum)
```

```python
#manupilating strings

# Iterate over each character in the string
for char in "Python Programming":
    print(char)
    # Check if the character is a vowel and convert to uppercase
    if char.lower() in 'aeiou':
        print(char.upper(), end=' ')
    else:
        # Print non-vowel characters as is
        print(char, end=' ')
```

```python
#printing a triangle

# Outer loop to control the number of rows
for i in range(1, 6):
    # Inner loop to print '*' i times
    for j in range(i):
        print('*', end=' ')
    # Print a newline after each row is completed
    print()
```

# While loop

The `while` loop in Python executes a set of statements as long as a condition is true. It's typically used when the number of iterations is not known beforehand.

### Key Features:
- **Conditional looping**: Continues execution until a certain condition becomes false.
- **Infinite loops**: Can potentially run forever if the condition never becomes false, hence careful condition management is necessary.
- **Similar control statements**: Like `for` loops, `while` loops can also utilize `break`, `continue`, and `else`.


```python
# Initialize the starting number
number = 5

# Loop as long as number is greater than 0
while number > 0:
    print(number)
    # Decrement the number by 1
    number -= 1
```

```python
# Initialize variables
word = ""

char = "a"

# Loop as long as the length of word is less than 5
while len(word) < 5:
    word += char
    # Change the character after each iteration
    char = chr(ord(char) + 1)
    
    print(word)
```

```python
# Initialize the counter
count = 0

# Loop until count reaches 10
while count < 10:
    print("Count:", count)
    # Increment the counter
    count += 1
```

```python
# Original string
original_str = "hello"

reversed_str = ""

index = len(original_str) - 1

# Loop through the string backwards
while index >= 0:
    reversed_str += original_str[index]
    # Move to the previous character
    index -= 1

print("Reversed String:", reversed_str)
```

```python
# Calculating factorial of a number
num = 5

factorial = 1

# Loop to calculate factorial
while num > 1:
    print(num)
    factorial *= num
    print(factorial)
    num -= 1

print(f"The factorial is {factorial}")
```

```python
import random

# Creating a number guessing game
target_number = random.randint(1, 10)

guess = None

# Loop until the user guesses correctly
while guess != target_number:
    
    guess = int(input("Guess a number between 1 and 10: "))
    
    if guess < target_number:
        
        print("Guess higher!")
        
    elif guess > target_number:
        
        print("Guess lower!")

print("Congratulations! You guessed the number.")
```

## Theory of Loop Control Commands in Python

Loop control commands in Python are crucial for altering the execution flow of loops (`for` or `while` loops). The three primary loop control commands are `break`, `continue`, and `else`.

1. **`break` Command**: 
   - **Purpose**: Used to immediately terminate the loop.
   - **Behavior**: When `break` is encountered, no further iterations of the loop are executed.
   - **Use Case**: Commonly used in scenarios where a condition outside of the regular iteration necessitates an early loop termination, such as finding a desired element.

2. **`continue` Command**: 
   - **Purpose**: Skips the rest of the code inside the loop for the current iteration, moving directly to the next iteration.
   - **Behavior**: The loop does not terminate but continues with the next iteration.
   - **Use Case**: Useful when certain conditions within the loop require skipping some parts of the loop body for specific iterations.

3. **`else` Clause with Loops**: 
   - **Purpose**: Executes a block of code once after the loop completes normally.
   - **Behavior**: The code in the `else` block is executed only if the loop finishes its iteration sequence without encountering a `break` statement.
   - **Use Case**: Useful for scenarios where you need to confirm that the loop completed without interruptions.


```python
# Example: Finding the first multiple of 5 in a range

for num in range(1, 20):
    
    if num % 5 == 0:
        
        print(f"First multiple of 5 found: {num}")
        
        break  # Break the loop when the first multiple of 5 is found
```

```python
# Example: Printing only odd numbers in a range and skipping even numbers
for num in range(1, 10):
    
    if num % 2 != 0:
        
        continue  # Skip the iteration if the number is even
        
    print(f"even Number: {num}")
```

```python
# Example: Checking for a prime number and using else for a confirmation message
num = 33
divisor = 2
is_prime = True

while divisor <= num / 2:
    
    if num % divisor == 0:
        
        is_prime = False
        
        print(f"{num} is not a prime number.")
        
        break  # Break the loop if a divisor is found
    print(divisor)
    divisor += 1
    
else:
    # This part executes if the loop wasn't broken, meaning the number is prime
    
    print(f"{num} is a prime number.")
```
