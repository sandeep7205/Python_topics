```python
# Example 1: A simple function to add two numbers
def add_numbers(a, b):
    """
    This function takes two numbers and returns their sum.
    Parameters:
    a (int or float): The first number.
    b (int or float): The second number.
    
    Returns:
    int or float: The sum of a and b.
    """
    return a + b

# Example usage of add_numbers
result = add_numbers(5, 3)


print("The sum is:", result)


```

```python
# Example 2: A function to check if a number is even or odd

def check_even_odd(number):
    """
    This function checks if a number is even or odd.
    Parameters:
    number (int): The number to check.
    
    Returns:
    str: "Even" if the number is even, "Odd" otherwise.
    """
    
    number=int(number)
    
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

# Example usage of check_even_odd
print("The number 7 is:", check_even_odd(7))


check_even_odd("3")
```

```python
# Example 3: A function to calculate the factorial of a number


def calculate_factorial(n):
    """
    This function calculates the factorial of a given number.
    Parameters:
    n (int): The number to calculate the factorial of.
    
    Returns:
    int: The factorial of n.
    """
    if n == 0 or n == 1:
        return 1
    else:
        return n * calculate_factorial(n-1)

# Example usage of calculate_factorial
print("The factorial of 5 is:", calculate_factorial(5))


```

```python
# Example 4: A function to compare two numbers
def compare_numbers(a, b):
    """
    This function compares two numbers and returns a relation between them.
    Parameters:
    a (int or float): The first number.
    b (int or float): The second number.
    
    Returns:
    str: "a is greater", "b is greater", or "a and b are equal".
    """
    if a > b:
        return "a is greater"
    elif a < b:
        return "b is greater"
    else:
        return "a and b are equal"

# Example usage of compare_numbers
print(compare_numbers(10, 5))
print(compare_numbers(2, 3))
print(compare_numbers(7, 7))
```
