# Scope of Functions

In programming, the *scope* of a variable defines the region of the code where a variable is accessible. Python has two basic scopes of variables: **global** and **local**. Understanding these scopes is crucial for managing variable values and their availability across different parts of your code.

## Global Variables

Global variables are declared outside of all functions. These variables can be accessed from any part of the program, including inside functions. However, to modify a global variable inside a function, you must explicitly declare it as `global` within the function. Otherwise, Python will treat it as a local variable.

### Example of Global Variable

```python
x = "global"

def access_global():
    print("Inside the function, x is: " + x)

access_global()
print("Outside the function, x is: " + x)


````python
x = "global"

def access_global():
    print("Inside the function, x is: " + x)

access_global()


print("Outside the function, x is: " + x)
````

## Local Variables

Local variables are declared inside a function and can only be accessed within that function. They are created when the function starts and are destroyed when the function returns. Local variables cannot be accessed outside of their function, providing a form of variable privacy.

````python
def function_scope():
    y = "local"
    print("Inside the function, y is: " + y)
    
    return y

function_scope()
# Trying to access y outside of its function will raise an error
# print(y)  # This would raise a NameError
````

````python
y=function_scope()
````
