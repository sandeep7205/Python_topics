```python
def convert_length(value, from_unit, to_unit):
    if from_unit == "meters" and to_unit == "feet":
        return value * 3.28084, "feet"
    elif from_unit == "feet" and to_unit == "meters":
        return value / 3.28084, "meters"
    else:
        return "Invalid length conversion"


```

```python
def convert_temperature(value, from_unit, to_unit):
    if from_unit == "Celsius" and to_unit == "Fahrenheit":
        return (value * 9/5) + 32, "Fahrenheit"
    elif from_unit == "Fahrenheit" and to_unit == "Celsius":
        return (value - 32) * 5/9, "Celsius"
    else:
        return "Invalid temperature conversion"

```

```python
def unit_converter():
    while True:
        conversion_type = input("Enter conversion type (length/temperature/quit): ")
        if conversion_type == "quit":
            print("Exiting converter.")
            break
        
        value = float(input("Enter value to convert: "))
        from_unit = input("To unit (meters/feet/Celsius/Fahrenheit): ")
        to_unit = input("From unit (meters/feet/Celsius/Fahrenheit): ")

        if conversion_type == "length":
            result, unit = convert_length(value, from_unit, to_unit)
        elif conversion_type == "temperature":
            result, unit = convert_temperature(value, from_unit, to_unit)
        else:
            print("Invalid conversion type.")
            continue
        
        print(f"Converted value: {result} {unit}")

```

```python
unit_converter()
```
