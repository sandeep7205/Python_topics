
# Solutions for Module 7.2 Data Structures - List Basics Coding Questions

This notebook contains solutions to the coding questions on list basics.


```
# Function to find the average sales
def average_sales(sales):
    sales_with_value = [sale for sale in sales if sale != 0]
    return sum(sales_with_value) / len(sales_with_value) if sales_with_value else 0
```

```
# Function to apply a discount to a list of prices
def apply_discount(prices, discount_percentage):
    return [price * (1 - discount_percentage / 100) for price in prices]
```

```
# Function to merge two sorted lists
def merge_sorted_lists(list1, list2):
    merged_list = []
    i, j = 0, 0
    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:
            merged_list.append(list1[i])
            i += 1
        else:
            merged_list.append(list2[j])
            j += 1
    merged_list.extend(list1[i:])
    merged_list.extend(list2[j:])
    return merged_list
```

```
# Function to remove all instances of a value from a list
def remove_value_from_list(the_list, value):
    return [element for element in the_list if element != value]
```

```
# Function to categorize and sum expenses
def categorize_expenses(expenses):
    categories = {}
    for category, amount in expenses:
        if category in categories:
            categories[category] += amount
        else:
            categories[category] = amount
    return categories
```

```
# Function to convert names to initials
def names_to_initials(names):
    return [''.join(name[0] for name in full_name.split()) for full_name in names]
```

```
# Function to filter out negative numbers
def filter_positive(temperatures):
    return [temp for temp in temperatures if temp > 0]
```

```
# Function to increase prices by a percentage
def increase_prices(products, percentage):
    for product in products:
        product['price'] += product['price'] * (percentage / 100)
```

```
# Function to count word frequencies
def word_frequencies(words):
    frequencies = {}
    for word in words:
        if word in frequencies:
            frequencies[word] += 1
        else:
            frequencies[word] = 1
    return frequencies
```

```
# Function to remove duplicates from a list
def remove_duplicates(the_list):
    unique = []
    for item in the_list:
        if item not in unique:
            unique.append(item)
    return unique
```
