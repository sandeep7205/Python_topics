
# Solutions for Module 7.3 Data Structures - List Questions Coding Questions

This notebook contains solutions to the coding questions on list questions.


```
# Function to filter out all orders below a certain amount
def filter_orders(orders, threshold):
    return [order for order in orders if order > threshold]
```

```
# Function to find the maximum score
def max_score(scores):
    return max(scores)
```

```
# Function to normalize a list of temperatures
def normalize_temperatures(temperatures):
    min_temp = min(temperatures)
    max_temp = max(temperatures)
    range_temp = max_temp - min_temp
    return [(temp - min_temp) / range_temp * 100 for temp in temperatures]
```

```
# Function to calculate the total bill and apply a discount
def calculate_bill(prices, discount):
    total = sum(prices)
    return total - (total * discount / 100)
```

```
# Function to shuffle a playlist
import random
def shuffle_playlist(playlist):
    random.shuffle(playlist)
    return playlist
```

```
# Function to filter words by length
def filter_words(words, n):
    return [word for word in words if len(word) > n]
```

```
# Function to remove 'N/A' or 'None' values
def clean_data(data):
    return [value for value in data if value not in ['N/A', 'None']]
```

```
# Function to merge two sorted lists into a single sorted list
def merge_sorted_lists(list1, list2):
    result = []
    while list1 and list2:
        if list1[0] < list2[0]:
            result.append(list1.pop(0))
        else:
            result.append(list2.pop(0))
    result.extend(list1 or list2)
    return result
```

```
# Function to categorize expenses
def categorize_expenses(expenses, thresholds):
    categories = {'High': [], 'Medium': [], 'Low': []}
    for expense in expenses:
        if expense > thresholds['High']:
            categories['High'].append(expense)
        elif expense > thresholds['Medium']:
            categories['Medium'].append(expense)
        else:
            categories['Low'].append(expense)
    return categories
```

```
# Function to generate a list of unique random numbers
import random
def unique_random_numbers(range_start, range_end, num_elements):
    numbers = set()
    while len(numbers) < num_elements:
        numbers.add(random.randint(range_start, range_end))
    return list(numbers)
```
