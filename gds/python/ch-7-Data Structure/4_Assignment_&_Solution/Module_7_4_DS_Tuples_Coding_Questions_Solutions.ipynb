
# Solutions for Module 7.4 Data Structures - Tuples Coding Questions

This notebook contains solutions to the coding questions on tuples.


```
# Function that returns the sum and average of numbers in a tuple
def sum_and_average(numbers):
    sum_numbers = sum(numbers)
    average = sum_numbers / len(numbers) if numbers else 0
    return sum_numbers, average
```

```
# Function to convert a list of names into a tuple of names that start with a certain letter
def filter_names_by_letter(names, letter):
    return tuple(name for name in names if name.startswith(letter))
```

```
# Function that scales tuple of coordinates by a factor
def scale_coordinates(coordinates, factor):
    return tuple(coordinate * factor for coordinate in coordinates)
```

```
# Function that merges two tuples of equal length into a tuple of pairs
def merge_tuples(tuple1, tuple2):
    return tuple(zip(tuple1, tuple2))
```

```
# Function to search for an element in a tuple and return its index
def find_element_index(elements, target):
    for index, element in enumerate(elements):
        if element == target:
            return index
    return -1
```

```
# Function to count the occurrences of all items in a tuple
def count_occurrences(elements):
    occurrences = {}
    for element in elements:
        if element in occurrences:
            occurrences[element] += 1
        else:
            occurrences[element] = 1
    return occurrences
```

```
# Function to calculate distances from the origin for points represented as tuples of tuples
from math import sqrt
def distances_from_origin(points):
    return tuple(sqrt(x**2 + y**2) for x, y in points)
```

```
# Function to filter out non-numeric types from a tuple
def filter_numeric_types(elements):
    return tuple(element for element in elements if isinstance(element, (int, float)))
```

```
# Function to concatenate tuple of strings with a separator
def concatenate_strings(strings, separator):
    return separator.join(strings)
```

```
# Function to filter points within a given radius from the origin
from math import sqrt
def points_within_radius(points, radius):
    return tuple(point for point in points if sqrt(point[0]**2 + point[1]**2) <= radius)
```
