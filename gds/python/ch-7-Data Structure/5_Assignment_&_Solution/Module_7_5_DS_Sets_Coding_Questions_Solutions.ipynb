
# Solutions for Module 7.5 Data Structures - Sets Coding Questions

This notebook contains solutions to the coding questions on sets.


```
# Function to remove duplicates from a list
def remove_duplicates(a_list):
    return set(a_list)
```

```
# Function to find common product IDs
def find_common_ids(set1, set2):
    return set1 & set2
```

```
# Function to determine if a set is a subset of another
def is_subset(set1, set2):
    return set1.issubset(set2)
```

```
# Function to return unique characters in a string
def unique_characters(a_string):
    return set(a_string)
```

```
# Function to find the union of skills required for projects
def union_of_skills(skill_sets):
    return set.union(*skill_sets)
```

```
# Function to compare two sets and return unique elements from the first
def unique_to_first_set(set1, set2):
    return set1 - set2
```

```
# Function to update a set with elements from a list
def update_set(a_set, a_list):
    a_set.update(a_list)
    return a_set
```

```
# Function to sanitize a set of names
def sanitize_names(names):
    return {name.strip().lower() for name in names}
```

```
# Function to filter email addresses by valid domains
def filter_emails(emails, valid_domains):
    return {email for email in emails if email.split('@')[-1] in valid_domains}
```

```
# Function to calculate the symmetric difference between two sets
def symmetric_difference(set1, set2):
    return set1 ^ set2
```
