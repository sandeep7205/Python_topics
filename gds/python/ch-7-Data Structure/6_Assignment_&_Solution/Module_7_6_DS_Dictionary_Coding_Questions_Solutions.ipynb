
# Solutions for Module 7.6 Data Structures - Dictionary Coding Questions

This notebook contains solutions to the coding questions on dictionaries.


```
# Function to count the frequency of words
def word_frequency(text):
    words = text.split()
    frequencies = {}
    for word in words:
        frequencies[word] = frequencies.get(word, 0) + 1
    return frequencies
```

```
# Function to merge two dictionaries
def merge_dictionaries(dict1, dict2):
    return {**dict1, **dict2}
```

```
# Function to invert a dictionary
def invert_dictionary(a_dict):
    return {value: key for key, value in a_dict.items()}
```

```
# Function to find the highest-paid employee
def highest_paid_employee(salary_dict):
    return max(salary_dict, key=salary_dict.get)
```

```
# Function to convert a list of tuples into a dictionary
def list_to_dict(orders):
    return {order[0]: order[1:] for order in orders}
```

```
# Function to filter a dictionary
def filter_dictionary(a_dict, threshold):
    return {key: value for key, value in a_dict.items() if value >= threshold}
```

```
# Function to apply a discount to product prices
def apply_discount(products, discount_percentage):
    return {product: price * (1 - discount_percentage / 100) for product, price in products.items()}
```

```
# Function to categorize words by their first letter
def categorize_words(words):
    categories = {}
    for word in words:
        letter = word[0]
        if letter in categories:
            categories[letter].append(word)
        else:
            categories[letter] = [word]
    return categories
```

```
# Function to invert courses and students
def invert_courses(courses):
    students = {}
    for course, names in courses.items():
        for name in names:
            if name in students:
                students[name].append(course)
            else:
                students[name] = [course]
    return students
```

```
# Function to find all nodes reachable from a given node
def reachable_nodes(graph, start_node):
    visited = set()
    stack = [start_node]
    while stack:
        node = stack.pop()
        if node not in visited:
            visited.add(node)
            stack.extend(graph[node] - visited)
    return list(visited)
```
