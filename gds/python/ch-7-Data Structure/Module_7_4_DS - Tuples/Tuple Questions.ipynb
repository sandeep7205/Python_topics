```python
# Q: How can you create a tuple with mixed data types and access its elements?
# A tuple containing an integer, a string, and a float
my_tuple = (1, "Hello", 3.14)
print(my_tuple[0])  # Access the first element (integer)
print(my_tuple[1])  # Access the second element (string)
```

```python
# Q: How to perform advanced tuple unpacking to separate head from the rest of the items?
# A tuple of numbers
numbers = (1, 2, 3,4,5,6)

# Advanced unpacking
first, middle, *last = numbers

print(f"First: {first}, Middle: {middle}, Last: {last}")
```

```python
# Q: How to define a function that returns multiple values and unpack those values?
def min_max_avg(numbers):
    
    
    return min(numbers), max(numbers), sum(numbers) / len(numbers)  # Returns a tuple

# Using the function and unpacking
minimum, maximum, average = min_max_avg([1, 2, 3, 4, 5])
print(f"Minimum: {minimum}, Maximum: {maximum}, Average: {average}")
```

```python
# Q: How to use tuples as keys in dictionaries and dynamically access and update values?

# Dictionary with tuple keys representing country and city

capitals = {("India", "Delhi"): "New Delhi", ("USA", "Washington"): "Washington, D.C."}

# Dynamically accessing and updating
capitals[("France", "Paris")] = "Paris"
for key, value in capitals.items():
    print(f"Country: {key[0]}, Capital City: {value}")
```

```python
# Q: Write a function that processes a list of tuples representing (number, weight) and calculates the weighted average.


def weighted_average(tuple_list):
    total_weight = sum(weight for _, weight in tuple_list)
    weighted_sum = sum(number * weight for number, weight in tuple_list)
    return weighted_sum / total_weight if total_weight > 0 else 0

# Using the function with a list of tuples
tuple_list = [(10, 2), (20, 1), (30, 3)]
print(f"Weighted Average: {weighted_average(tuple_list)}")
```
