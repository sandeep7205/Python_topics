```python
# Q: Write a function that merges two dictionaries and returns the resulting dictionary.


def merge_dictionaries(dict1, dict2):
    
    """Merge two dictionaries. In case of overlapping keys, values from dict2 overwrite those of dict1."""
    merged_dict = dict1.copy()
    
    # Create a copy of the first dictionary
    
    
    merged_dict.update(dict2) 
    
    # Update with the second dictionary
    return merged_dict

# Example usage
dict1 = {'a': 1, 'b': 2}
dict2 = {'b': 3, 'c': 4}
print(merge_dictionaries(dict1, dict2))
```

```python
# Q: Write a function that inverts the keys and values of a dictionary. Assume all values are hashable and unique.
def invert_dictionary(original_dict):
    """Invert a dictionary's keys and values."""
    inverted_dict = {value: key for key, value in original_dict.items()}
    return inverted_dict

# Example usage
my_dict = {'a': 1, 'b': 2, 'c': 3}
print(invert_dictionary(my_dict))
```

```python
# Q: Write a function to filter a dictionary by keys that start with a certain string.


def filter_dict_by_key_start(original_dict, start_str):
    """Filter the dictionary to include only keys that start with start_str."""
    filtered_dict = {key: value for key, value in original_dict.items() if key.startswith(start_str)}
    return filtered_dict

# Example usage
my_dict = {'apple': 1, 'banana': 2, 'apricot': 3, 'orange': 4}
print(filter_dict_by_key_start(my_dict, 'a'))
```

```python
# Q: Write a function that counts the occurrences of each item in a list and returns a dictionary with items as keys and their counts as values.
def count_items(item_list):
    """Count occurrences of each item in the given list."""
    count_dict = {}
    
    for item in item_list:
        if item in count_dict:
            count_dict[item] += 1
        else:
            count_dict[item] = 1
    return count_dict

# Example usage
items = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']
print(count_items(items))
```
