This notebook provides answers to theory questions about the introduction to classes in object-oriented programming.


## Question 1: Concept of a Class vs. Object
A class in object-oriented programming (OOP) is a blueprint for creating objects. It defines a set of attributes (fields or properties) and methods (functions or procedures) that are common to all objects of that class. An object, on the other hand, is an instance of a class. It represents a specific entity that is created based on the class blueprint, having its own set of attribute values and capable of executing the methods defined in the class.

## Question 2: Encapsulation
Encapsulation is a fundamental concept in OOP that involves bundling the data (attributes) and the methods (functions) that operate on the data into a single unit called a class. It helps to restrict the access to some of the object's components, which makes the misuse of the object more difficult and ensures the internal representation of the object is hidden from the outside. This also facilitates modularity and maintenance.

## Question 3: Classes in Modeling Real-World Problems
Classes help in modeling real-world problems by providing a way to create abstractions of real-world entities in software development. For example, classes can be used to represent employees, products, or services in a system. Each class can encapsulate the specific attributes and behaviors relevant to the entity it represents, making it easier to understand, manage, and use these entities within programs.

## Question 4: Superclass and Subclass Relationship
The relationship between a superclass and a subclass is one of inheritance in OOP. A subclass inherits or extends the functionality of a superclass. It can use the attributes and methods of the superclass as well as introduce its own or override the superclass's methods. This relationship enables code reuse and the creation of more specific classes from general ones.

## Question 5: Role of Constructors
Constructors are special methods used in classes to initialize new objects. When an object is created, the constructor is automatically called to set the initial state of the object by assigning values to its attributes. Constructors ensure that the objects are properly created with valid initial values.

## Question 6: Polymorphism Example
Polymorphism in OOP allows methods in different classes to have the same name but different implementations. For example, a `draw()` method might be used in different shape classes like Circle, Square, and Triangle. Each class implements `draw()` in a way specific to drawing the shape it represents, allowing the method to be used polymorphically at runtime.

## Question 7: Method Overriding
Method overriding is a feature of OOP that allows a subclass to provide a specific implementation of a method that is already defined in its superclass. This is used when the subclass needs to alter or enhance the behavior of the superclass method. It ensures that the version of the method that is most relevant to the current object is executed.

## Question 8: Access Modifiers
Access modifiers (public, private, protected) define the visibility of class members (attributes and methods). Public members are accessible from anywhere, private members can only be accessed within the class itself, and protected members are accessible within the class and by subclasses. They control how the data and behavior of a class are exposed and help in implementing encapsulation.

## Question 9: Use of Static Methods and Variables
Static methods and variables are associated with the class itself, rather than any specific instance of the class. They are used when some property or behavior is shared across all instances of the class, such as a method for validating email addresses or a counter to keep track of the number of objects created. Static members can be accessed without creating an instance of the class.

## Question 10: Composition vs. Inheritance
Composition and inheritance are two approaches to code reuse in OOP. Composition involves building classes from other classes by including them as fields, while inheritance is about creating a new class that inherits attributes and behaviors from an existing class. Composition is chosen over inheritance when you need to model a has-a relationship rather than an is-a relationship, offering more flexibility and easier maintenance.
