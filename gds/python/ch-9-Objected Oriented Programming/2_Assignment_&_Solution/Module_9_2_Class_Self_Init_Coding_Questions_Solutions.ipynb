
# Solutions for Module 9.2 Class - Self & Init: Coding Questions

This notebook contains solutions to coding questions related to class initialization and the self parameter in object-oriented programming.


```
# Class for a bank account
class BankAccount:
    def __init__(self):
        self.balance = 0
    
    def deposit(self, amount):
        self.balance += amount
    
    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
        else:
            print("Insufficient funds.")
    
    def display_balance(self):
        print(f"Account balance: ${self.balance}")
```

```
# Class to represent a product
class Product:
    def __init__(self, name, price, inventory_quantity):
        self.name = name
        self.price = price
        self.inventory_quantity = inventory_quantity
    
    def display_product_details(self):
        print(f"Product: {self.name}, Price: ${self.price}, Inventory: {self.inventory_quantity}")
```

```
# Class representing a Book
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
    
    def display_book_info(self):
        print(f"Title: {self.title}, Author: {self.author}, Pages: {self.pages}")
```

```
# Class for managing a to-do list
class ToDoList:
    def __init__(self):
        self.tasks = []
    
    def add_task(self, task):
        self.tasks.append(task)
    
    def remove_task(self, task):
        if task in self.tasks:
            self.tasks.remove(task)
        else:
            print("Task not found.")
    
    def display_tasks(self):
        for task in self.tasks:
            print(task)
```

```
# Class for handling user profiles
class UserProfile:
    def __init__(self, user_name, bio):
        self.user_name = user_name
        self.bio = bio
        self.posts = []
    
    def add_post(self, post):
        self.posts.append(post)
    
    def delete_post(self, post):
        if post in self.posts:
            self.posts.remove(post)
        else:
            print("Post not found.")
    
    def display_profile(self):
        print(f"User: {self.user_name}, Bio: {self.bio}")
        print("Posts:")
        for post in self.posts:
            print(post)
```

```
# Class to represent a vehicle
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
    
    def display_description(self):
        print(f"{self.year} {self.make} {self.model}")
```

```
# Class for a simple inventory system
class Inventory:
    def __init__(self):
        self.items = {}
    
    def add_item(self, name, quantity):
        if name in self.items:
            self.items[name] += quantity
        else:
            self.items[name] = quantity
    
    def remove_item(self, name, quantity):
        if name in self.items and self.items[name] >= quantity:
            self.items[name] -= quantity
            if self.items[name] == 0:
                del self.items[name]
        else:
            print("Insufficient quantity or item not found.")
```

```
# Class that represents a circle
import math
class Circle:
    def __init__(self, radius):
        self.radius = radius
    
    def calculate_area(self):
        return math.pi * self.radius ** 2
```

```
# Class for an email system
class EmailSystem:
    def __init__(self):
        self.emails = []
    
    def compose_email(self, email):
        self.emails.append(email)
    
    def send_email(self, email):
        print(f"Sending email: {email}")
    
    def delete_email(self, email):
        if email in self.emails:
            self.emails.remove(email)
        else:
            print("Email not found.")
```

```
# Class to manage a movie collection
class MovieCollection:
    def __init__(self):
        self.movies = []
    
    def add_movie(self, movie):
        self.movies.append(movie)
    
    def find_movie(self, title):
        for movie in self.movies:
            if movie['title'] == title:
                return movie
        return None
    
    def list_movies(self):
        for movie in self.movies:
            print(movie)
```
