```python

# Question 1: Implement a class for a shopping cart

class ShoppingCart:
    def __init__(self):
        # Initialize an empty dictionary to store items and their prices
        self.items = {}
    
    def add_item(self, item, price):
        # Add an item with its price to the cart
        self.items[item] = price
    
    def remove_item(self, item):
        # Remove an item from the cart if it exists
        if item in self.items:
            del self.items[item]
    
    def calculate_total(self):
        # Calculate the total cost of items in the cart
        return sum(self.items.values())

# Example usage:
cart = ShoppingCart()
cart.add_item("Apple", 1.5)
cart.add_item("Banana", 2.0)
cart.remove_item("Apple")
total = cart.calculate_total()
print(total)  # Output: 2.0


# Question 2: Create a class for managing a digital library

class DigitalLibrary:
    def __init__(self):
        # Initialize an empty list to store book titles
        self.books = []
    
    def add_book(self, title):
        # Add a book title to the library
        self.books.append(title)
    
    def remove_book(self, title):
        # Remove a book title from the library if it exists
        if title in self.books:
            self.books.remove(title)
    
    def list_books(self):
        # List all book titles in the library
        return self.books

# Example usage:
library = DigitalLibrary()
library.add_book("1984")
library.add_book("To Kill a Mockingbird")
library.remove_book("1984")
books = library.list_books()
print(books)  # Output: ['To Kill a Mockingbird']


# Question 3: Design a class representing a calculator

class Calculator:
    def add(self, a, b):
        # Return the sum of a and b
        return a + b
    
    def subtract(self, a, b):
        # Return the difference between a and b
        return a - b
    
    def multiply(self, a, b):
        # Return the product of a and b
        return a * b
    
    def divide(self, a, b):
        # Return the quotient of a divided by b, with error handling for division by zero
        if b != 0:
            return a / b
        else:
            return "Error: Division by zero"

# Example usage:
calc = Calculator()
print(calc.add(10, 5))       # Output: 15
print(calc.divide(10, 0))    # Output: Error: Division by zero


# Question 4: Write a class for a user profile in an application

class UserProfile:
    def __init__(self, name, age, email):
        # Initialize user profile attributes
        self.name = name
        self.age = age
        self.email = email
    
    def update_profile(self, name=None, age=None, email=None):
        # Update user profile attributes if new values are provided
        if name:
            self.name = name
        if age:
            self.age = age
        if email:
            self.email = email
    
    def display_profile(self):
        # Display user profile information
        print(f"Name: {self.name}, Age: {self.age}, Email: {self.email}")

# Example usage:
user = UserProfile("John Doe", 30, "john.doe@example.com")
user.update_profile(age=31)
user.display_profile()  # Output: Name: John Doe, Age: 31, Email: john.doe@example.com


# Question 5: Develop a class for a game character with methods to modify the character's stats

class GameCharacter:
    def __init__(self, health, strength, agility):
        # Initialize character stats
        self.health = health
        self.strength = strength
        self.agility = agility
    
    def modify_stat(self, stat, amount):
        # Modify a specific stat by a given amount, ensuring it doesn't go below 0
        if stat == "health":
            self.health = max(0, self.health + amount)
        elif stat == "strength":
            self.strength = max(0, self.strength + amount)
        elif stat == "agility":
            self.agility = max(0, self.agility + amount)
    
    def display_stats(self):
        # Display the character's stats
        print(f"Health: {self.health}, Strength: {self.strength}, Agility: {self.agility}")

# Example usage:
character = GameCharacter(100, 50, 30)
character.modify_stat("health", -20)
character.display_stats()  # Output: Health: 80, Strength: 50, Agility: 30


# Question 6: Implement a class for a simple email system

class EmailSystem:
    def __init__(self):
        # Initialize lists to store draft and sent emails
        self.drafts = []
        self.sent = []
    
    def compose_email(self, email):
        # Add an email to the drafts list
        self.drafts.append(email)
    
    def send_email(self, email):
        # Move an email from the drafts list to the sent list if it exists in drafts
        if email in self.drafts:
            self.drafts.remove(email)
            self.sent.append(email)
    
    def delete_email(self, email):
        # Remove an email from either the drafts or sent list if it exists
        if email in self.drafts:
            self.drafts.remove(email)
        elif email in self.sent:
            self.sent.remove(email)

# Example usage:
email_system = EmailSystem()
email_system.compose_email("Draft 1")
email_system.send_email("Draft 1")
email_system.delete_email("Draft 1")
print(email_system.drafts)  # Output: []
print(email_system.sent)    # Output: []


# Question 7: Create a class for a bank system

class BankSystem:
    def __init__(self):
        # Initialize a dictionary to store account information
        self.accounts = {}
    
    def create_account(self, account_number):
        # Create a new account with a balance of 0
        self.accounts[account_number] = 0
    
    def deposit(self, account_number, amount):
        # Deposit an amount to an account if it's valid
        if amount > 0 and account_number in self.accounts:
            self.accounts[account_number] += amount
    
    def withdraw(self, account_number, amount):
        # Withdraw an amount from an account if it's valid and there's sufficient balance
        if amount > 0 and account_number in self.accounts and self.accounts[account_number] >= amount:
            self.accounts[account_number] -= amount
    
    def check_balance(self, account_number):
        # Return the balance of an account
        return self.accounts.get(account_number, "Account not found")

# Example usage:
bank = BankSystem()
bank.create_account("123")
bank.deposit("123", 1000)
bank.withdraw("123", 500)
balance = bank.check_balance("123")
print(balance)  # Output: 500


# Question 8: Design a class for a temperature converter

class TemperatureConverter:
    def fahrenheit_to_celsius(self, fahrenheit):
        # Convert Fahrenheit to Celsius
        return (fahrenheit - 32) * 5/9
    
    def celsius_to_fahrenheit(self, celsius):
        # Convert Celsius to Fahrenheit
        return (celsius * 9/5) + 32
    
    def celsius_to_kelvin(self, celsius):
        # Convert Celsius to Kelvin
        return celsius + 273.15
    
    def kelvin_to_celsius(self, kelvin):
        # Convert Kelvin to Celsius
        return kelvin - 273.15

# Example usage:
converter = TemperatureConverter()
print(converter.fahrenheit_to_celsius(100))  # Output: 37.77777777777778


# Question 9: Write a class for managing tasks in a to-do list application

class ToDoList:
    def __init__(self):
        # Initialize an empty list to store tasks
        self.tasks = []
    
    def add_task(self, task):
        # Add a task to the list
        self.tasks.append(task)
    
    def complete_task(self, task):
        # Remove a task from the list if it exists
        if task in self.tasks:
            self.tasks.remove(task)
    
    def display_tasks(self):
        # Display all tasks in the list
        return self.tasks

# Example usage:
todo = ToDoList()
todo.add_task("Write code")
todo.complete_task("Write code")
tasks = todo.display_tasks()
print(tasks)  # Output: []


# Question 10: Develop a class for an online store's product

class Product:
    def __init__(self, name, price, stock):
        # Initialize product attributes
        self.name = name
        self.price = price
        self.stock = stock
    
    def apply_discount(self, discount):
        # Apply a discount to the product's price
        if 0 <= discount <= 1:
            self.price *= (1 - discount)
    
    def restock(self, amount):
        # Increase the product's stock by a given amount
        if amount > 0:
            self.stock += amount
    
    def display_product_info(self):
        # Display product information
        print(f"Name: {self.name}, Price: {self.price}, Stock: {self.stock}")



    
```
