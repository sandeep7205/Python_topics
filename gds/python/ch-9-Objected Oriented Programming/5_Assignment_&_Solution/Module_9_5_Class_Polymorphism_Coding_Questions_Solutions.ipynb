
# Solutions for Module 9.5 Class - Polymorphism: Coding Questions

This notebook contains solutions to coding questions related to class polymorphism in object-oriented programming.


```python
# Question 1: Implement a base class for vehicles with derived classes

class Vehicle:
    def display_details(self):
        raise NotImplementedError("Subclasses should implement this method")

class Car(Vehicle):
    def display_details(self):
        return "This is a car."

class Truck(Vehicle):
    def display_details(self):
        return "This is a truck."

class Motorcycle(Vehicle):
    def display_details(self):
        return "This is a motorcycle."

# Example usage:
vehicles = [Car(), Truck(), Motorcycle()]
for vehicle in vehicles:
    print(vehicle.display_details())
# Output: 
# This is a car.
# This is a truck.
# This is a motorcycle.


# Question 2: Create a function that computes the total area of different shapes

class Shape:
    def area(self):
        raise NotImplementedError("Subclasses should implement this method")

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14 * self.radius ** 2

class Square(Shape):
    def __init__(self, side):
        self.side = side
    
    def area(self):
        return self.side ** 2

def total_area(shapes):
    return sum(shape.area() for shape in shapes)

# Example usage:
shapes = [Circle(2), Square(3), Circle(1)]
print(total_area(shapes))  # Output: 19.56


# Question 3: Design a class hierarchy for animals with a method to make a sound

class Animal:
    def make_sound(self):
        raise NotImplementedError("Subclasses should implement this method")

class Dog(Animal):
    def make_sound(self):
        return "Woof!"

class Cat(Animal):
    def make_sound(self):
        return "Meow!"

class Bird(Animal):
    def make_sound(self):
        return "Chirp!"

def animal_sounds(animals):
    for animal in animals:
        print(animal.make_sound())

# Example usage:
animals = [Dog(), Cat(), Bird()]
animal_sounds(animals)
# Output:
# Woof!
# Meow!
# Chirp!


# Question 4: Write a class for a digital media player with different file formats

class MediaFile:
    def play(self):
        raise NotImplementedError("Subclasses should implement this method")

class MP3(MediaFile):
    def play(self):
        return "Playing MP3 file."

class WAV(MediaFile):
    def play(self):
        return "Playing WAV file."

class AAC(MediaFile):
    def play(self):
        return "Playing AAC file."

# Example usage:
files = [MP3(), WAV(), AAC()]
for file in files:
    print(file.play())
# Output:
# Playing MP3 file.
# Playing WAV file.
# Playing AAC file.


# Question 5: Develop a class structure for a messaging app

class MessagingChannel:
    def send_message(self, message):
        raise NotImplementedError("Subclasses should implement this method")

class EmailChannel(MessagingChannel):
    def send_message(self, message):
        return f"Sending email: {message}"

class SMSChannel(MessagingChannel):
    def send_message(self, message):
        return f"Sending SMS: {message}"

class InAppChannel(MessagingChannel):
    def send_message(self, message):
        return f"Sending in-app message: {message}"

# Example usage:
channels = [EmailChannel(), SMSChannel(), InAppChannel()]
for channel in channels:
    print(channel.send_message("Hello!"))
# Output:
# Sending email: Hello!
# Sending SMS: Hello!
# Sending in-app message: Hello!


# Question 6: Implement classes for different types of employees

class Employee:
    def calculate_bonus(self):
        raise NotImplementedError("Subclasses should implement this method")

class Manager(Employee):
    def calculate_bonus(self):
        return "Manager bonus calculated."

class Engineer(Employee):
    def calculate_bonus(self):
        return "Engineer bonus calculated."

class Salesperson(Employee):
    def calculate_bonus(self):
        return "Salesperson bonus calculated."

# Example usage:
employees = [Manager(), Engineer(), Salesperson()]
for employee in employees:
    print(employee.calculate_bonus())
# Output:
# Manager bonus calculated.
# Engineer bonus calculated.
# Salesperson bonus calculated.


# Question 7: Create classes representing geometric shapes to calculate perimeter

class GeometricShape:
    def perimeter(self):
        raise NotImplementedError("Subclasses should implement this method")

class Rectangle(GeometricShape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def perimeter(self):
        return 2 * (self.width + self.height)

class Triangle(GeometricShape):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    
    def perimeter(self):
        return self.a + self.b + self.c

class CircleShape(GeometricShape):
    def __init__(self, radius):
        self.radius = radius
    
    def perimeter(self):
        return 2 * 3.14 * self.radius

def largest_perimeter(shapes):
    return max(shapes, key=lambda shape: shape.perimeter())

# Example usage:
shapes = [Rectangle(3, 4), Triangle(3, 4, 5), CircleShape(2)]
largest = largest_perimeter(shapes)
print(largest.perimeter())  # Output: 15


# Question 8: Write classes for different types of payment

class PaymentMethod:
    def process_payment(self, amount):
        raise NotImplementedError("Subclasses should implement this method")

class CreditCard(PaymentMethod):
    def process_payment(self, amount):
        return f"Processing credit card payment of {amount}"

class PayPal(PaymentMethod):
    def process_payment(self, amount):
        return f"Processing PayPal payment of {amount}"

class BankTransfer(PaymentMethod):
    def process_payment(self, amount):
        return f"Processing bank transfer of {amount}"

# Example usage:
payments = [CreditCard(), PayPal(), BankTransfer()]
for payment in payments:
    print(payment.process_payment(100))
# Output:
# Processing credit card payment of 100
# Processing PayPal payment of 100
# Processing bank transfer of 100


# Question 9: Design a class hierarchy for a file system

class BaseFile:
    def display_contents(self):
        raise NotImplementedError("Subclasses should implement this method")

class File(BaseFile):
    def __init__(self, name, content):
        self.name = name
        self.content = content
    
    def display_contents(self):
        return f"File: {self.name}, Content: {self.content}"

class Directory(BaseFile):
    def __init__(self, name):
        self.name = name
        self.contents = []
    
    def add_item(self, item):
        self.contents.append(item)
    
    def display_contents(self):
        return f"Directory: {self.name}, Contents: {[item.display_contents() for item in self.contents]}"

# Example usage:
file1 = File("file1.txt", "Hello World")
file2 = File("file2.txt", "Python")
directory = Directory("MyFolder")
directory.add_item(file1)
directory.add_item(file2)
print(directory.display_contents())
# Output:
# Directory: MyFolder, Contents: ['File: file1.txt, Content: Hello World', 'File: file2.txt, Content: Python']


# Question 10: Develop classes for a UI framework with different UI elements

class UIElement:
    def draw(self):
        raise NotImplementedError("Subclasses should implement this method")

class Button(UIElement):
    def draw(self):
        return "Drawing a button"

class Slider(UIElement):
    def draw(self):
        return "Drawing a slider"

class TextBox(UIElement):
    def draw(self):
        return "Drawing a text box"

# Example usage:
elements = [Button(), Slider(), TextBox()]
for element in elements:
    print(element.draw())
# Output:
# Drawing a button
# Drawing a slider
# Drawing a text box
```
