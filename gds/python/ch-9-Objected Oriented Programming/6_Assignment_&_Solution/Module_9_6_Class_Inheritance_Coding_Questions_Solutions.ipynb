
# Solutions for Module 9.6 Class - Inheritance: Coding Questions

This notebook contains solutions to coding questions related to class inheritance in object-oriented programming.


```python
# Question 1: Design a base class for Vehicles and derive classes for Car, Truck, and Motorcycle

class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def display_details(self):
        return f"{self.year} {self.make} {self.model}"

class Car(Vehicle):
    def __init__(self, make, model, year, num_doors):
        super().__init__(make, model, year)
        self.num_doors = num_doors

class Truck(Vehicle):
    def __init__(self, make, model, year, cargo_capacity):
        super().__init__(make, model, year)
        self.cargo_capacity = cargo_capacity

class Motorcycle(Vehicle):
    def __init__(self, make, model, year, has_sidecar):
        super().__init__(make, model, year)
        self.has_sidecar = has_sidecar

# Example usage:
car = Car("Toyota", "Camry", 2020, 4)
truck = Truck("Ford", "F-150", 2018, 10000)
motorcycle = Motorcycle("Harley", "Sportster", 2019, False)
print(car.display_details())
print(truck.display_details())
print(motorcycle.display_details())
# Output:
# 2020 Toyota Camry
# 2018 Ford F-150
# 2019 Harley Sportster


# Question 2: Create a class hierarchy for a university system

class Person:
    def __init__(self, name, address):
        self.name = name
        self.address = address

class Student(Person):
    def __init__(self, name, address, student_id):
        super().__init__(name, address)
        self.student_id = student_id
    
    def enroll(self, course):
        return f"{self.name} enrolled in {course}"

class Instructor(Person):
    def __init__(self, name, address, employee_id):
        super().__init__(name, address)
        self.employee_id = employee_id
    
    def teach(self, course):
        return f"{self.name} teaches {course}"

# Example usage:
student = Student("John Doe", "123 Main St", "S12345")
instructor = Instructor("Dr. Smith", "456 College Ave", "I67890")
print(student.enroll("Math 101"))
print(instructor.teach("Math 101"))
# Output:
# John Doe enrolled in Math 101
# Dr. Smith teaches Math 101


# Question 3: Implement a base class for Electronic Devices

class ElectronicDevice:
    def __init__(self, brand, power_on):
        self.brand = brand
        self.power_on = power_on
    
    def toggle_power(self):
        self.power_on = not self.power_on
        return "Powered on" if self.power_on else "Powered off"

class Mobile(ElectronicDevice):
    def __init__(self, brand, power_on, battery_life):
        super().__init__(brand, power_on)
        self.battery_life = battery_life

class Laptop(ElectronicDevice):
    def __init__(self, brand, power_on, is_gaming):
        super().__init__(brand, power_on)
        self.is_gaming = is_gaming

# Example usage:
mobile = Mobile("Samsung", False, 24)
laptop = Laptop("Dell", True, True)
print(mobile.toggle_power())
print(laptop.toggle_power())
# Output:
# Powered on
# Powered off


# Question 4: Develop a game with a base class Character

class Character:
    def __init__(self, name, health):
        self.name = name
        self.health = health

    def attack(self):
        raise NotImplementedError("Subclasses should implement this method")

class Wizard(Character):
    def __init__(self, name, health, mana):
        super().__init__(name, health)
        self.mana = mana
    
    def attack(self):
        return f"{self.name} casts a spell"

class Knight(Character):
    def __init__(self, name, health, armor):
        super().__init__(name, health)
        self.armor = armor
    
    def attack(self):
        return f"{self.name} swings a sword"

class Archer(Character):
    def __init__(self, name, health, arrows):
        super().__init__(name, health)
        self.arrows = arrows
    
    def attack(self):
        return f"{self.name} shoots an arrow"

# Example usage:
wizard = Wizard("Gandalf", 100, 50)
knight = Knight("Arthur", 150, 30)
archer = Archer("Robin", 80, 20)
print(wizard.attack())
print(knight.attack())
print(archer.attack())
# Output:
# Gandalf casts a spell
# Arthur swings a sword
# Robin shoots an arrow


# Question 5: Write a class hierarchy for a restaurant system

class MenuItem:
    def __init__(self, name, price):
        self.name = name
        self.price = price

class DrinkItem(MenuItem):
    def __init__(self, name, price, is_alcoholic):
        super().__init__(name, price)
        self.is_alcoholic = is_alcoholic

class FoodItem(MenuItem):
    def __init__(self, name, price, is_vegan):
        super().__init__(name, price)
        self.is_vegan = is_vegan

# Example usage:
drink = DrinkItem("Coke", 2.5, False)
food = FoodItem("Salad", 5.0, True)
print(drink.name, drink.price, drink.is_alcoholic)
print(food.name, food.price, food.is_vegan)
# Output:
# Coke 2.5 False
# Salad 5.0 True


# Question 6: Design a base class Shape

class Shape:
    def area(self):
        raise NotImplementedError("Subclasses should implement this method")

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height

class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height
    
    def area(self):
        return 0.5 * self.base * self.height

# Example usage:
circle = Circle(3)
rectangle = Rectangle(4, 5)
triangle = Triangle(6, 7)
print(circle.area())
print(rectangle.area())
print(triangle.area())
# Output:
# 28.26
# 20
# 21.0


# Question 7: Create a class hierarchy for a transportation system

class Transport:
    def __init__(self, speed, capacity):
        self.speed = speed
        self.capacity = capacity

class LandTransport(Transport):
    def __init__(self, speed, capacity, num_wheels):
        super().__init__(speed, capacity)
        self.num_wheels = num_wheels
    
    def drive(self):
        return f"Driving at {self.speed} km/h with {self.num_wheels} wheels"

class WaterTransport(Transport):
    def __init__(self, speed, capacity, type_of_boat):
        super().__init__(speed, capacity)
        self.type_of_boat = type_of_boat
    
    def sail(self):
        return f"Sailing at {self.speed} knots on a {self.type_of_boat}"

# Example usage:
car = LandTransport(100, 4, 4)
boat = WaterTransport(30, 10, "yacht")
print(car.drive())
print(boat.sail())
# Output:
# Driving at 100 km/h with 4 wheels
# Sailing at 30 knots on a yacht


# Question 8: Implement a base class for all types of bank accounts

class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        return self.balance

class CheckingAccount(BankAccount):
    def __init__(self, balance, overdraft_limit):
        super().__init__(balance)
        self.overdraft_limit = overdraft_limit

class SavingsAccount(BankAccount):
    def __init__(self, balance, interest_rate):
        super().__init__(balance)
        self.interest_rate = interest_rate
    
    def add_interest(self):
        self.balance += self.balance * self.interest_rate
        return self.balance

# Example usage:
checking = CheckingAccount(1000, 500)
savings = SavingsAccount(2000, 0.05)
print(checking.deposit(500))
print(savings.add_interest())
# Output:
# 1500
# 2100


# Question 9: Design a base class Product

class Product:
    def __init__(self, product_id, name, price):
        self.product_id = product_id
        self.name = name
        self.price = price

class Perishable(Product):
    def __init__(self, product_id, name, price, expiration_date):
        super().__init__(product_id, name, price)
        self.expiration_date = expiration_date
    
    def is_expired(self, current_date):
        return current_date > self.expiration_date

class NonPerishable(Product):
    def __init__(self, product_id, name, price):
        super().__init__(product_id, name, price)

# Example usage:
milk = Perishable(1, "Milk", 1.5, "2024-07-30")
rice = NonPerishable(2, "Rice", 10)
print(milk.is_expired("2024-08-01"))
print(rice.name)
# Output:
# True
# Rice


# Question 10: Develop a content management system class hierarchy

class Content:
    def __init__(self, title, creator):
        self.title = title
        self.creator = creator
    
    def display(self):
        raise NotImplementedError("Subclasses should implement this method")

class Video(Content):
    def __init__(self, title, creator, duration):
        super().__init__(title, creator)
        self.duration = duration
    
    def display(self):
        return f"Playing video: {self.title} by {self.creator}, duration: {self.duration} mins"

class Article(Content):
    def __init__(self, title, creator, word_count):
        super().__init__(title, creator)
        self.word_count = word_count
    
    def display(self):
        return f"Reading article: {self.title} by {self.creator}, word count: {self.word_count}"

class Podcast(Content):
    def __init__(self, title, creator, length):
        super().__init__(title, creator)
        self.length = length
    
    def display(self):
        return f"Listening to podcast: {self.title} by {self.creator}, length: {self.length} mins"

# Example usage:
video = Video("Python Tutorial", "John Doe", 45)
article = Article("Learning Python", "Jane Smith", 1500)
podcast = Podcast("Python in Data Science", "Alice Johnson", 30)
print(video.display())
print(article.display())
print(podcast.display())
# Output:
# Playing video: Python Tutorial by John Doe, duration: 45 mins
# Reading article: Learning Python by Jane Smith, word count: 1500
# Listening to podcast: Python in Data Science by Alice Johnson, length: 30 mins
```
