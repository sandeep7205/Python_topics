# What is a Variable? 

VARIABLES are entities which help us store information and retrieve it later.
A variable with a fixed name can store information of nature like numeric, textual, boolean etc.
A Python variable is a reserved memory location to store values. In other words, a variable in a python program gives data to the computer for processing.
The type of data contained in a variable can be changed at user's will.

You can store numbers in variables.
The standard rule is you write the variable name followed by = sign and the value it will take

##  Rules for naming a variable in Python

* Variables names must start with a letter or an underscore like _ product , product_


* The remainder of your variable name may consist of letters, numbers and underscores


* spacy1, pyThon,machine_learning are some valid variable names


* Names are case sensitive.


* case_sensitive, CASE_SENSITIVE, and Case_Sensitive are each a different variable.

* Names can not contain spaces, use _ instead

      :'",<>/?|\!@#%^&*~-+
      
      
* It is considered best practice that names are lowercase with underscores


* Avoid using Python built-in keywords like `list`, `str` , `def` etc. We will talk more about such conventions later on

## Different data types we will encounter in Python
* ### <u>Numeric</u>  - Numeric variables take values which are numbers like 9, 3.14, 0, Inf


* ### <u>String </u>  - String variables are used to store textual information


* ### <u>Boolean</u>  - Boolean variables have two modes either True or False. A definite judge of statements!

# <u> Integers and Floats </u>

## Let's talk about numbers!

* A lot many different types of numbers are supported in Python like integers (int type), real numbers (float type), complex numbers. We will mostly use integer and floating point numbers.


* Integers are just whole numbers, positive or negative. For example: 2 and -2 are examples of integers.


* Floating point numbers in Python are notable because they have a decimal point in them, or use an exponential (E) to define the number. For example 2.0 and -2.1 are examples of floating point numbers. 4E2 (4 times 10 to the power of 2) is also an example of a floating point number in Python.


* In computing, floating-point arithmetic is arithmetic using formulaic representation of real numbers as an approximation to support a trade-off between range and precision. You can always control the number of digits coming after the decimal, hence they are called floating-point numbers

```python
# Examples of functions applicable to both int and float


# abs() - absolute value
print(abs(-5))  # Returns 5
print(abs(-3.2))  # Returns 3.2



# round() - round to specified decimal places
print(round(3.756))  # Returns 4
print(round(3.756, 1))  # Returns 3.8




# pow() - power of a number
print(pow(2, 3))  # Returns 8
print(pow(4.5, 2))  # Returns 20.25




# Conversion between int and float
print(int(3.5))  # Returns 3
print(float(5))  # Returns 5.0



# Examples of functions specific to float

# is_integer() - checks if float is a whole number
print((3.0).is_integer())  # Returns True
print((3.5).is_integer())  # Returns False



# as_integer_ratio() - represents the float as a fraction
print((3.5).as_integer_ratio())  # Returns (7, 2)



# hex() - converts the float to a hexadecimal string
print((3.5).hex())  # Returns a string like '0x1.c000000000000p+1'



# fromhex() - converts a hexadecimal string back to a float
print(float.fromhex('0x3.a000000000000p0'))  # Returns 3.625



# Examples of functions specific to int

# bit_length() - number of bits to represent an integer in binary
print((1024).bit_length())  # Returns 11



# to_bytes() and from_bytes() - for byte representation
print((1024).to_bytes(2, byteorder='big'))  # Converts int to bytes
print(int.from_bytes(b'\x04\x00', byteorder='big'))  # Converts bytes back to int

```

# <u>Boolean Variables </u>

* A Boolean variable only takes two values either True or False. It is used for comparisons

```python
# Define some variables
a = 10
b = 20

# Using comparison operators to create Boolean variables
is_equal = (a == b)  # Checks if a is equal to b
is_not_equal = (a != b)  # Checks if a is not equal to b
is_less_than = (a < b)  # Checks if a is less than b
is_greater_than = (a > b)  # Checks if a is greater than b
is_less_than_or_equal = (a <= b)  # Checks if a is less than or equal to b
is_greater_than_or_equal = (a >= b)  # Checks if a is greater than or equal to b

# Print the results
print("Is a equal to b:", is_equal)
print("Is a not equal to b:", is_not_equal)
print("Is a less than b:", is_less_than)
print("Is a greater than b:", is_greater_than)
print("Is a less than or equal to b:", is_less_than_or_equal)
print("Is a greater than or equal to b:", is_greater_than_or_equal)
```

```python
# Example 1: Comparing Values
is_equal = (5 == 5)  # Checks if 5 equals 5
print("5 == 5 is", is_equal)



is_greater = (10 > 5)  # Checks if 10 is greater than 5
print("10 > 5 is", is_greater)




# Example 2: Logical Operations
is_true = True
is_false = False

result_and = is_true and is_false  # Logical AND
print("True and False is", result_and)

result_or = is_true or is_false  # Logical OR


print("True or False is for OR", result_or)

result_not = not is_true  # Logical NOT
print("Not True is", result_not)




# Example 3: Boolean in Functions
def is_even(number):
    return number % 2 == 0

print("Is 4 even?", is_even(4))
print("Is 5 even?", is_even(5))

# Example 4: Booleans in Loops
print("Counting with while loop:")
counter = 0
while counter < 5:
    print(counter)
    counter += 1
```
