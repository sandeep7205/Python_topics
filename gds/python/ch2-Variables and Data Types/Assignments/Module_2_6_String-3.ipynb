Question 1: Finding a Substring

How do you find the first occurrence of the substring "Python" in the string "Learning Python with Python tutorials"?

Hint: Use the method that searches for the substring and returns the lowest index where it begins.

```python
full_str = "Learning Python with Python tutorials"
print("1st occurance of python - ", full_str.find("Python"))
```

Question 2: Counting Substring Occurrences

How many times does the substring "is" appear in the string "This is an island"?

Hint: There's a string method that returns the count of non-overlapping occurrences of a substring.

```python
str = "This is an island"
print("'is' appeared - ", str.count('is'))
```

Question 3: Checking String Start

Determine if the string "Python programming is fun." starts with "Python".

Hint: Use the string method that checks if a string starts with a specified prefix.

```python
str = "Python programming is fun."

print("starts with Python - ", str.startswith("Python"))
```

Question 4: Reversing Words 

Reverse the order of words in the sentence "Hello World Python" to "Python World Hello" without reversing the words themselves.

Hint: Split the string into words, reverse the list, then join it back into a string.

```python
str = "Hello World Python"
all_words = str.split(" ")
print(all_words)

reversed_words = " ".join(all_words[::-1])

# reversed_words = " ".join(reversed(all_words))

# all_words.reverse()  # This reverses the list in place
# reversed_words = " ".join(all_words)  # Joins the reversed list into a string

print(reversed_words)
```

Question 5: Modifying Letter Case

Convert "hello world" to "Hello World" using a string method to capitalize the first letter of each word.

Hint: Look for a method that transforms the first character of each word to uppercase and the rest to lowercase.

```python
str = 'hello world'
print(str.capitalize()) #Converts the first character to upper case
print(str.title()) #Converts the first character of each word to upper case
```

Question 6: Removing Leading Characters

Remove all leading zeros from the string "0001234" to get "1234".

Hint: Use the method that trims characters (not just whitespace) from the beginning of the string.

Question 7: Checking Digit Characters

Question 8: Swapping Letter Case

Swap the case of all letters in the string "Python 3.9" (uppercase to lowercase and vice versa).

Hint: Find the string method that inverts the case of all letters in the string.

```python
str = "Python 3.9"
print("swapping the cases from [", str, "] to [", str.swapcase(), "]")
```

Question 9: Joining with a Separator

Create a comma-separated string from the list ["apple", "banana", "cherry"].

Hint: Use the string method that concatenates a collection of strings with a specified separator between elements.

```python
list = ["apple", "banana", "cherry"]
print(list, " - after join the list - ", ", ".join(list))
```

Question 10: Stripping Substring from Ends

Given a string "xxHello Worldxx", remove the 'xx' from both the beginning and the end of the string.

Hint: Apply the method that returns a copy of the string with specified leading and trailing characters removed.

```python
str = "xxHello Worldxx"
print("left trim str - ", str.lstrip("x"))
print("right trim str - ", str.rstrip("x"))
print("full trim str - ", str.strip("x"))
```
