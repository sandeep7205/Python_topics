Question 3: Slicing a Substring

Extract the word Hello from the string Hello, world.

Hint: Use slicing by specifying the start and end indices of the substring you want to extract.

```python
my_str = "Hello, World"
print("Slicing a string - ", my_str[0:5])
print("Slicing a string - ", my_str[:5])
```

Question 5: Skipping Characters in a Slice

If you want to extract every other letter from abcdefghijk starting from a, how would you do that?

Hint: The step parameter in slicing controls how many characters to skip.

```python
my_str = "abcdefghijk"
print("skip odd charcters - ", my_str[1::2])
print("skip even charcters - ", my_str[::2])
```

Question 6: Default Values in Slicing

What happens if you omit the start or the end value while slicing a string, such as [:5] or [3:] in Hello, world?

Hint: Omitting start or end values in slices uses default values that represent the beginning or the end of the string, respectively.

```python
my_str = "Hello, World"
print("First five characters - ", my_str[:5])
print("After the third character - ", my_str[3:])
```

Question 8: Length of a Slice

Without directly counting, how can you determine the length of the slice Hello from Hello, world?

Hint: Consider using the len() function on your slice.

```python
my_str = "Hello, World"
sub_str = my_str[:my_str.find(",")]
print(sub_str)
length_of_sub_str = len(sub_str)
print("Length of 'Hello': ", length_of_sub_str)




hello = "Hello world"[:5]
length_of_hello = len(hello)
print("Length of 'Hello':", length_of_hello)
```
