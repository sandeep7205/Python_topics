# Introduction to Code Indentation

Indentation refers to the spaces at the beginning of a code line. In Python, indentation is used to define the block of code. All statements with the same distance to the right belong to the same block of code, also known as the code suite. If the indentation is not correct, the code will not execute as expected, and Python will raise an `IndentationError`.

## Why Indentation Matters

- **Readability**: Proper indentation makes the code more readable and understandable.
- **Structure**: It indicates the start and end of control structures like loops and functions.
- **Necessity in Python**: Unlike other languages that use braces to show where blocks of code begin and end, Python uses indentation as a part of the syntax.


Common Practices for Indentation
Use 4 spaces per indentation level.
Be consistent with the use of tabs or spaces (Python 3 disallows mixing the use of tabs and spaces for indentation).
Coding Example: Using Indentation in a Loop

Correct indentation in a for loop


```python
for i in range(5):
    print("Number is:", i)
    if i % 2 == 0:
        print("It's an even number")
    else:
        print("It's an odd number")
```

Use Case: Indentation in Conditional Statements and Functions
Imagine we want to write a function that categorizes a person's age group:


By using proper indentation, we clearly define the different conditions in our function.

```python
def categorize_age(age):
    if age < 12:
        return 'Child'
    elif age < 18:
        return 'Teenager'
    elif age < 60:
        return 'Adult'
    else:
        return 'Senior'
```
