🔹 let's use HTML & CSS to colorize the text 

```python
from IPython.display import display, HTML

colors = ["red", "yellow", "green", "cyan", "blue", "magenta", "orange", "purple", "lime", "pink", "gold", "aqua"]
text = "Sandeep Kumar Mishra"

colored_text = "".join(
    f'<span style="color:{colors[i % len(colors)]}; font-size:50px; font-weight:bold;">{char}</span>'
    for i, char in enumerate(text)
)

display(HTML(colored_text))
```
