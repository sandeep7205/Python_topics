```python
# 🧠 Concept Recap:
# In Python:
#   (a // b) gives the quotient (result of floor division)
#   (a % b) gives the remainder
# And they always satisfy this equation:
#   (a // b) * b + (a % b) = a

# Floor division means rounding the result of a/b down (towards negative infinity)
# NOT just "truncating" or "rounding toward zero".

a = 5
b = 2

# Normal division gives a float (decimal) result
print(f"division of {a} / {b} = ", a / b)  
# → 5 / 2 = 2.5  (normal division, gives exact decimal value)

# Floor division (//) gives the quotient by rounding down
print(f"quotient(floor division) of {a} // {b} = ", a // b)   
# → 5 // 2 = 2   (because floor(2.5) = 2, largest integer ≤ 2.5)

# When 'a' is negative, floor division rounds the result DOWN (toward -∞)
print(f"quotient(floor division) of -{a} // {b} = ", -a // b)
# → (-5) / 2 = -2.5 → floor(-2.5) = -3
# So, (-5) // 2 = -3 (NOT -2) because floor(-2.5) = -3

# When 'b' is negative, sign of result flips
print(f"quotient(floor division) of {a} // -{b} = ", a // -b)
# → 5 / -2 = -2.5 → floor(-2.5) = -3

# Both negative → positive result (negative ÷ negative = positive)
print(f"quotient(floor division) of -{a} // -{b} = ", -a // -b)
# → (-5) / (-2) = 2.5 → floor(2.5) = 2

# ------------------------------
# 🧩 Now, remainders using '%'
# Formula:  a % b = a - (a // b) * b
# ------------------------------

print(f"remainder of {a} % {b} = ", a % b)
# → a // b = 2, so remainder = 5 - (2*2) = 5 - 4 = 1

print(f"remainder of -{a} % {b} = ", -a % b)
# → (-5) // 2 = -3, so remainder = -5 - (-3*2) = -5 - (-6) = 1
# ✅ Remainder always has the same sign as the divisor (b), which is +2 here, so remainder = +1

print(f"remainder of {a} % -{b} = ", a % -b)
# → 5 // -2 = -3, so remainder = 5 - (-3 * -2) = 5 - 6 = -1
# ✅ Remainder has the same sign as divisor (b = -2), so remainder = -1

print(f"remainder of -{a} % -{b} = ", -a % -b)
# → (-5) // (-2) = 2, so remainder = -5 - (2 * -2) = -5 - (-4) = -1
# ✅ Remainder sign = divisor sign (b = -2 → remainder = -1)

# ------------------------------
# ✅ Golden Rule to Remember:
# (a // b) * b + (a % b) == a   → always True in Python
# ------------------------------
```
