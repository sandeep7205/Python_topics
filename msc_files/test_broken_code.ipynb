```python
# 🧠 Concept Recap:
# In Python:
#   (a // b) gives the quotient (result of floor division)
#   (a % b) gives the remainder
# And they always satisfy this equation:
#   (a // b) * b + (a % b) = a

# Floor division means rounding the result of a/b down (towards negative infinity)
# NOT just "truncating" or "rounding toward zero".

a = 5
b = 2

# Normal division gives a float (decimal) result
print(f"division of {a} / {b} = ", a / b)  
# → 5 / 2 = 2.5  (normal division, gives exact decimal value)

# Floor division (//) gives the quotient by rounding down
print(f"quotient(floor division) of {a} // {b} = ", a // b)   
# → 5 // 2 = 2   (because floor(2.5) = 2, largest integer ≤ 2.5)

# When 'a' is negative, floor division rounds the result DOWN (toward -∞)
print(f"quotient(floor division) of -{a} // {b} = ", -a // b)
# → (-5) / 2 = -2.5 → floor(-2.5) = -3
# So, (-5) // 2 = -3 (NOT -2) because floor(-2.5) = -3

# When 'b' is negative, sign of result flips
print(f"quotient(floor division) of {a} // -{b} = ", a // -b)
# → 5 / -2 = -2.5 → floor(-2.5) = -3

# Both negative → positive result (negative ÷ negative = positive)
print(f"quotient(floor division) of -{a} // -{b} = ", -a // -b)
# → (-5) / (-2) = 2.5 → floor(2.5) = 2

# ------------------------------
# 🧩 Now, remainders using '%'
# Formula:  a % b = a - (a // b) * b
# ------------------------------

print(f"remainder of {a} % {b} = ", a % b)
# → a // b = 2, so remainder = 5 - (2*2) = 5 - 4 = 1

print(f"remainder of -{a} % {b} = ", -a % b)
# → (-5) // 2 = -3, so remainder = -5 - (-3*2) = -5 - (-6) = 1
# ✅ Remainder always has the same sign as the divisor (b), which is +2 here, so remainder = +1

print(f"remainder of {a} % -{b} = ", a % -b)
# → 5 // -2 = -3, so remainder = 5 - (-3 * -2) = 5 - 6 = -1
# ✅ Remainder has the same sign as divisor (b = -2), so remainder = -1

print(f"remainder of -{a} % -{b} = ", -a % -b)
# → (-5) // (-2) = 2, so remainder = -5 - (2 * -2) = -5 - (-4) = -1
# ✅ Remainder sign = divisor sign (b = -2 → remainder = -1)

# ------------------------------
# ✅ Golden Rule to Remember:
# (a // b) * b + (a % b) == a   → always True in Python
# ------------------------------
```

```python
def multiply(a: int, b: int) -> int:
    """
    Multiplies two non-negative integers (a * b) using only addition and recursion.
    """
    if a == 0 or b == 0:
        return 0
        
    # Optimization: Use the smaller number as the counter to minimize recursive calls.
    base = max(a, b)
    counter = min(a, b)
    
    def helper_add(current_base, count):
        # Base Case: When count hits 0, stop and return 0
        if count == 0:
            return 0
        
        # Recursive Step: Add base to the result of (count - 1)
        return current_base + helper_add(current_base, count - 1)

    return helper_add(base, counter)

# --- Example ---
final_result = multiply(5, 3)
print(f"Result: {final_result}")
```

```python
def multiply(a: int, b: int) -> int:
    """
    Multiplies two non-negative integers (a * b) using only addition and recursion.
    
    NOTE: This simple version is for educational purposes and only works for a >= 0 and b >= 0.
    It includes a tree-style visualization to show the execution flow.
    """
    if a == 0 or b == 0:
        return 0
        
    # Optimization: Use the smaller number as the counter to minimize recursive calls.
    base = max(a, b)
    counter = min(a, b)
    
    print(f"\n--- Visualizing Execution: {a} * {b} ---")
    print(f"Strategy: Add {base} exactly {counter} times.")
    print("Execution Tree:")
    
    def helper_add(current_base, count, depth=0):
        # Create a graphical tree structure using pipes for indentation
        prefix = "│   " * depth
        
        # Base Case: When count hits 0, stop and return 0
        if count == 0:
            print(f"{prefix}└── [Base Case] Count is 0 -> Return 0")
            return 0
        
        # Recursive Step: Print what we are about to do
        print(f"{prefix}├── [Step {depth+1}] Need {current_base} + result of (count={count-1})")
        
        # --- DIVE DOWN: Call the function recursively ---
        sub_result = helper_add(current_base, count - 1, depth + 1)
        
        # --- BUBBLE UP: Calculate total for this level ---
        total = current_base + sub_result
        
        # Show the value returning up the chain
        print(f"{prefix}└── [Return] {current_base} + {sub_result} = {total}")
        return total

    return helper_add(base, counter)

# --- Visualization Example ---
# This will print a tree showing how 5 is added 3 times
final_result = multiply(50, 10)
print(f"\nFinal Calculated Result: {final_result}")
```
