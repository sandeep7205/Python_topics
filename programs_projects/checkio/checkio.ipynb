# Rectangle Perimeter
> This function should take two positive numbers (length and width) as inputs and return the perimeter of a rectangle.


How it’s used:
 - in architectural and engineering applications for calculating the perimeter of buildings or rooms;
 - in computer graphics to calculate the perimeter of a rectangle on a screen.



````python
def rectangle_perimeter(length: int, width: int) -> int:   
    print (f"rectangle_perimeter for length = {length} and width = {width} is [{(length + width)*2}]")
    return (length + width)*2

assert rectangle_perimeter(2, 4) == 12
assert rectangle_perimeter(3, 5) == 16
assert rectangle_perimeter(10, 20) == 60
assert rectangle_perimeter(7, 2) == 18
# assert rectangle_perimeter(2, 5) == 10 # when an assert passes, it doesn't produce any output. It only raises an AssertionError if the condition fails.
````

# Integer Sign Determination
> Identify whether a given integer is positive, negative, or zero and return a respective string: "positive", "negative" or "zero".

> How it’s used: understanding the sign of a number can be useful in financial software, scientific simulations, and many algorithms to determine subsequent logic.


````python
def determine_sign(num: int) -> str:
    if num > 0:
        return "positive"
    elif num < 0:
        return "negative"
    elif num == 0:
        return "zero"
        
    # return 'positive' if num > 0 else 'negative' if num < 0 else 'zero'
    # return "zero" * int(num == 0) + "positive" * int(num > 0) + "negative" * int(num < 0)

print("Example:")
print(determine_sign(11))

# These "asserts" are used for self-checking
assert determine_sign(5) == "positive"
assert determine_sign(0) == "zero"
assert determine_sign(-10) == "negative"
assert determine_sign(1) == "positive"
assert determine_sign(-1) == "negative"
assert determine_sign(999) == "positive"
assert determine_sign(-1000) == "negative"
assert determine_sign(123456789) == "positive"
assert determine_sign(-987654321) == "negative"
assert determine_sign(2) == "positive"
````

````python
def find_remainder(dividend: int, divisor: int) -> int:
    return (dividend % divisor)

def find_remainder1(dividend: int, divisor: int) -> int:
    
    if dividend < divisor:
        return dividend
        
    return find_remainder(dividend - divisor, divisor)

def find_remainder2(dividend: int, divisor: int) -> int:
    a = dividend - divisor
    if a < 0:
        return dividend
    if a < divisor:
        return a
    return find_remainder(a, divisor)

print("Example:")
# print(find_remainder(3, 2))
print(find_remainder1(27, 4))
# print(find_remainder2(3, 2))

# These "asserts" are used for self-checking
# assert find_remainder(10, 3) == 1
# assert find_remainder1(14, 4) == 2
# assert find_remainder2(27, 4) == 3
# assert find_remainder(10, 5) == 0
# assert find_remainder1(10, 1) == 0
# assert find_remainder2(5, 7) == 5
# assert find_remainder(7, 5) == 2
````

# Backward/Reverse String


````python
def backward_string(val: str) -> str:
    return val[::-1] # python slicing (sequence[start:stop:step])

def backward_string2(val: str) -> str:
    reversed_string = ""
    for char in val:
        reversed_string = char + reversed_string
    return reversed_string 

def backward_string3(val: str) -> str:
    return "".join(reversed(val)) 

print("Example:")
print(backward_string("Sandeep KUmar Mishra"))
print(backward_string2("Sandeep KUmar Mishra"))
print(backward_string3("Sandeep KUmar Mishra"))

# These "asserts" are used for self-checking
assert backward_string("val") == "lav"
assert backward_string("") == ""
assert backward_string("ohho") == "ohho"
assert backward_string("123456789") == "987654321"


#lambda
backward_string1 = lambda val: val[::-1]
print(backward_string1("Sandeep KUmar Mishra"))
````

# Just Fizz!

> You should write a function that will receive a positive integer and return: "Fizz" if the number is divisible by 3 (3, 6, 9 ...) otherwise convert the given number to a string (2 -> "2").


---

```python

# Method 1: Using ternary (conditional) operator
return str(num) if num % 3 else 'Fizz'

# Explanation:
# If num is NOT divisible by 3 (i.e., num % 3 != 0), this is True → return str(num)
# If num IS divisible by 3 (i.e., num % 3 == 0), this is False → return 'Fizz'
# Why? Because in Python, 0 is considered False, and any non-zero number is True


# Method 2: Using tuple indexing with boolean
return ('Fizz', str(num))[bool(num % 3)]

# Explanation: 
# ('Fizz', str(num)) - Create a tuple with two options: 
#   Index 0 → 'Fizz'
#   Index 1 → number converted to string

# [bool(num % 3)] - Then we use bool(num % 3) to decide which index to pick: [bool(num % 3)]
# If num is divisible by 3 → num % 3 == 0 → bool(0) → False → index 0 → 'Fizz'
# If not divisible → num % 3 != 0 → bool(1 or 2) → True → index 1 → str(num)

```
---


````python
def checkio(num: int) -> str:
    if (num%3) == 0:
        return "Fizz"
    else:
        return str(num)

def checkio1(num: int) -> str:
    return 'Fizz' if num % 3 == 0 else str(num)

def checkio2(num: int) -> str:
    return str(num) if num % 3 else 'Fizz'  

def checkio3(num: int) -> str:
    return ('Fizz', str(num))[bool(num%3)]

checkio = lambda num: ['Fizz',str(num)][bool(num%3)] 

print("Example:")
print(checkio(3))
print(checkio(7))

# These "asserts" are used for self-checking
assert checkio(15) == "Fizz"
assert checkio1(6) == "Fizz"
assert checkio2(1000) == "1000"
assert checkio3(999) == "Fizz"
````

# Number Length
> You have a integer. Try to find out how many digits it has.

````python
def number_length(value: int) -> int:
    return len(str(value)) if value >= 0 else len(str(abs(value)))
    # Convert the number to a string and return its length
    # If the number is non-negative, just convert and count digits
    # If it's negative, convert the absolute value to string and count digits (ignores minus sign)

print("Example:")
print(number_length(666))
print(number_length(-979796))

# These "asserts" are used for self-checking
assert number_length(10) == 2
assert number_length(0) == 1
assert number_length(4) == 1
assert number_length(-44) == 2
````
