# Rectangle Perimeter
> This function should take two positive numbers (length and width) as inputs and return the perimeter of a rectangle.


How it’s used:
 - in architectural and engineering applications for calculating the perimeter of buildings or rooms;
 - in computer graphics to calculate the perimeter of a rectangle on a screen.



```python
def rectangle_perimeter(length: int, width: int) -> int:   
    print (f"rectangle_perimeter for length = {length} and width = {width} is [{(length + width)*2}]")
    return (length + width)*2

assert rectangle_perimeter(2, 4) == 12
assert rectangle_perimeter(3, 5) == 16
assert rectangle_perimeter(10, 20) == 60
assert rectangle_perimeter(7, 2) == 18
# assert rectangle_perimeter(2, 5) == 10 # when an assert passes, it doesn't produce any output. It only raises an AssertionError if the condition fails.
```

# Is Even
> Check if the given number is even or not. Your function should return True if the number is even, and False if the number is odd.

```python
def is_even(num: int) -> bool:
    return True if num % 2 == 0 else False
    # return num % 2 == 0
    # return num & 1 == 0
    # return bin(num)[-1]=='0'
    # return not(num & 1)

assert is_even(2) == True
assert is_even(5) == False
assert is_even(0) == True
```

# Integer Sign Determination
> Identify whether a given integer is positive, negative, or zero and return a respective string: "positive", "negative" or "zero".

> How it’s used: understanding the sign of a number can be useful in financial software, scientific simulations, and many algorithms to determine subsequent logic.


```python
def determine_sign(num: int) -> str:
    if num > 0:
        return "positive"
    elif num < 0:
        return "negative"
    elif num == 0:
        return "zero"
        
    # return 'positive' if num > 0 else 'negative' if num < 0 else 'zero'
    # return "zero" * int(num == 0) + "positive" * int(num > 0) + "negative" * int(num < 0)

print("Example:")
print(determine_sign(11))

# These "asserts" are used for self-checking
assert determine_sign(5) == "positive"
assert determine_sign(0) == "zero"
assert determine_sign(-10) == "negative"
assert determine_sign(1) == "positive"
assert determine_sign(-1) == "negative"
assert determine_sign(999) == "positive"
assert determine_sign(-1000) == "negative"
assert determine_sign(123456789) == "positive"
assert determine_sign(-987654321) == "negative"
assert determine_sign(2) == "positive"
```

# Find Remainder
### Determine the remainder from division of two positive integers.


 > Input: Two integers (int): dividend (the number to be divided) and divisor (the number by which division is going to be performed).

 > How it’s used: remainder calculations are common in algorithms, especially in modular arithmetic which is frequently used in cryptography and computer graphics

```python
def find_remainder(dividend: int, divisor: int) -> int:
    return (dividend % divisor)

#---

def find_remainder1(dividend: int, divisor: int) -> int:
    
    if dividend < divisor:
        return dividend
        
    return find_remainder(dividend - divisor, divisor)
                
#---

def find_remainder2(dividend: int, divisor: int) -> int:
    a=dividend-divisor
    if a < 0:
        return dividend
    if a < divisor:
        return a
    return find_remainder(a,divisor)

print("Example:")
print(find_remainder(3, 2))
print(find_remainder1(3, 2))
print(find_remainder2(3, 2))

# These "asserts" are used for self-checking
assert find_remainder(10, 3) == 1
assert find_remainder1(14, 4) == 2
assert find_remainder2(27, 4) == 3
assert find_remainder(10, 5) == 0
assert find_remainder1(10, 1) == 0
assert find_remainder2(5, 7) == 5
assert find_remainder(7, 5) == 2
```
