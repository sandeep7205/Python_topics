### Area of a Circle

The formula for the area of a circle is:

### Area = 𝜋 (π (pi) ≈ 3.14159) × r2

```python
import math
import sys

# Function to get validated numeric input from user
def get_input(shape, param_name):
    while True:
        input_data = input(f"🔢 Enter the {shape.capitalize()} {param_name.capitalize()} (or type 'q' or 'no' to exit): ")
        
        # Allow user to exit anytime
        if input_data.lower() in ["q", "no"]:
            print("👋 Exiting the program. Thank you!")
            sys.exit()
        
        # Try to convert input to float
        try:
            value = float(input_data)
            if value < 0:
                print("⚠️ Please enter a positive number.")
            else:
                return value
        except ValueError:
            print("❌ Invalid input. Please enter numbers only.")

# Function to calculate and display area of a circle
def circle_area():
    # Get valid radius input from user
    radius = get_input('circle', 'radius')
    
    # Calculate area using πr² and round it to 2 decimal places
    area = round(math.pi * (radius ** 2), 2)
    
    # Print formatted result
    print(f"\n✅ The area of the circle with radius {radius} is {area} square units.\n")

# Allowable responses to repeat the process
repeat_responses = ["yes", "y"]

# Main program loop
try:
    while True:
        # Run the calculation
        circle_area()

        # Ask if the user wants to repeat
        again_input = input("🔁 Do you want to calculate another area? (yes/no): ").lower()

        # Exit if the answer is not in allowed responses
        if again_input not in repeat_responses:
            print("👋 Exiting the program. Stay curious, keep coding!")
            break

# Handle Ctrl+C gracefully
except KeyboardInterrupt:
    print("\n👋 Program interrupted. Goodbye!")
    sys.exit()
```
