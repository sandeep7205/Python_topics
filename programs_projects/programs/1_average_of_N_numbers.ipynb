# Find average of N numbers

> **Average = (Sum of N numbers) / N**

```python
# 📌 Define a list of numbers: 98, 79, 56, 34, 72
# These could represent test scores, values, etc.

# ✅ Step 1: Calculate the sum of the numbers manually
sum_of_numbers = (98 + 79 + 56 + 34 + 72)

# 🖨️ Step 2: Print the total sum
print(f"Sum: {sum_of_numbers}")

# ✅ Step 3: Calculate the average
# Total number of elements = 5
average = (sum_of_numbers / 5)

# 🖨️ Step 4: Print the average value
print(f"Average: {average}")
```

> Take N numbers from the user → Calculate their sum → Calculate their average → Show results.

```python
# 📌 This program calculates the sum and average of a user-defined number of integers using a while loop.

# ✅ Function to get valid integer input from the user
def input_num(i):
    while True:
        try:
            number = int(input(f"Enter number {i}: "))
            return number
        except ValueError:
            print("❌ Invalid input. Please enter an integer.")

# 🔢 Ask the user how many numbers they want to input
number_count = int(input("Enter Count: "))
print(f"🧮 Total numbers to input: {number_count}", flush=True)

# 🛠️ Initialize variables
i = 1                # Loop counter
sum = 0              # To store the sum of entered numbers
average = 0          # To store the calculated average

# 🔁 Use a while loop to collect inputs and calculate the sum
while i <= number_count:
    number = input_num(i)                      # Get valid number input
    print(f"✅ Number {i}: {number}", flush=True)          # Show entered number
    sum += number                              # Add to sum
    i += 1                                     # Move to next number

# 📊 After loop ends, calculate average
else:
    average = sum / number_count               # Calculate average

# 🖨️ Print the final results
print(f"\n🧮 Sum: {sum}")
print(f"📊 Average: {round(average, 2)}")

```

```python
# 📌 This program calculates the sum and average of a user-defined number of integers using a for loop.

def input_num(i):
    while True:
        try:
            number = int(input(f"Enter number {i}: "))
            return number
        except ValueError:
            print("❌ Invalid input. Please enter an integer.")

# Initialize total sum
total = 0

# Ask user how many numbers to input
number_count = int(input("Enter how many numbers you want to input: "))
print(f"\n📋 Count: {number_count}")

# Loop through the count and collect inputs
for i in range(1, number_count + 1):
    number = input_num(i)
    print(f"✅ number {i}: {number}")
    total += number

# Calculate average
if number_count > 0:
    average = total / number_count
else:
    average = 0

# Display final results
print(f"\n🧮 Sum: {total}")
print(f"📊 Average: {round(average, 2)}")
```

```python
# ✅ Input numbers in comma-separated format and calculate sum and average
# Example input: 56,uy,78,12.   3427,::,       96,==, 45, 78, 90, 100

# Prompt the user to enter numbers separated by commas
numbers = input("Enter input in comma-separated format: ")

# Show the raw input received
print(f"\n📝 Given Input: {numbers}")

# Split the input string into a list of values using commas
raw_inputs = numbers.split(",")

# Initialize variables to track total sum and valid count
total = 0
count = 0
valid_numbers = []  # List to store valid numbers

# Loop through each element from the input
for num in raw_inputs:
    cleaned = num.strip()  # Remove leading/trailing spaces

    try:
        # Try converting the cleaned input to a float
        value = float(cleaned)
        total += value             # Add to the total sum
        count += 1                 # Increase the valid number count
        valid_numbers.append(value)  # Save the valid number
    except ValueError:
        # If conversion fails, show a warning and skip the value
        print(f"⚠️ '{cleaned}' is not a valid number and will be skipped.")

# After processing all inputs, check if there were any valid numbers
if count == 0:
    print("\n❌ No valid numbers were entered.")
else:
    # Calculate average
    average = total / count

    # Display the results
    print(f"\n✅ Valid Numbers: {valid_numbers}")
    print(f"🔢 Count: {count}")
    print(f"🧮 Sum: {total}")
    print(f"📊 Average: {round(average, 2)}")
```
