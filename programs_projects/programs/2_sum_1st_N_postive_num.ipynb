# Sum of the first N positive integers

```python
#> Sum = (N * (N + 1))/2

number = int(input("Enter your number: "))
print(f"\nNumber : {number}")

sum = (number * (number + 1))/2

print(f"\nSum : {sum}")
```

```python
number = int(input("Enter your number: "))
print(f"\nNumber : {number}")

sum = 0
for i in range(0, (number+1)):
    sum += i
print(f"\nSum : {sum}")
```

```python
try:
    number = int(input("Enter your number: "))
    print(f"\nNumber : {number}")
    sum = 0
    i = 0
    while i  <= number:
        print(f"i : {i}")
        sum += i
        i += 1
    print(f"\nSum : {sum}")
except ValueError:
    print("Please enter a valid integer")

```
