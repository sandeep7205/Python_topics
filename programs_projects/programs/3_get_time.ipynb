# How to get time of a Python program's execution

```python
import time

# Display all available functions and attributes in the time module
print(f"Time module functions: {dir(time)}")

# Print the current time in a human-readable format
print(f"Current time: {time.ctime()}") 

# Pause the program for 0.5 seconds
print(f"Sleep time for 0.5 seconds: {time.sleep(0.5)}")

# ---------------------- Local Time Section ----------------------
print("\n\n----------Local time Module----------")

# Get current local time as a time.struct_time object
local_time = time.localtime()

# Print the full local time structure
print(f"Current Local time: {local_time}")

# Manually format and print the current local time
print(f"NOW : {local_time.tm_year}-{local_time.tm_mon}-{local_time.tm_mday} {local_time.tm_hour}:{local_time.tm_min}:{local_time.tm_sec}")

# Use strftime to format local time into a readable string
print(f"Formatted Local time with custom format: {time.strftime('%Y-%m-%d %H:%M:%S', local_time)}")

print("-------------------------------------\n\n")

# ---------------------- UTC Time Section ----------------------
print("\n\n----------UTC time Module----------")

# Get current UTC (Coordinated Universal Time)
utc_time = time.gmtime()

# Print the full UTC time structure
print(f"Current UTC time: {utc_time}")

# Manually format and print the UTC time
print(f"NOW : {utc_time.tm_year}-{utc_time.tm_mon}-{utc_time.tm_mday} {utc_time.tm_hour}:{utc_time.tm_min}:{utc_time.tm_sec}")

# Use strftime to format UTC time into a readable string
print(f"Formatted UTC time with custom format: {time.strftime('%Y-%m-%d %H:%M:%S', utc_time)}")

print("-------------------------------------\n\n")












# ---------------------- Execution Time Measurement ----------------------

# Record the start time in seconds
start_time = time.time()
print(f"Start Time: {start_time}")  # Print start timestamp

i = 0
user_input = int(input("Press Enter to End the loop..."))  # Wait for user input to end the loop

try:
    # Run a loop from 0 to 10, pausing 1 second between iterations
    while i <= user_input:
        print(f"Value of i: {i}")
        time.sleep(0.5)  # Pause for 0.5 seconds
        i += 1

    # Record the end time in seconds
    end_time = time.time()
    print(f"End Time: {end_time}")  # Print end timestamp

    # Calculate the total duration
    diff_time = end_time - start_time

    # Print total time in various units
    print(f"Total Time: {diff_time}")  
    print(f"Total Time in seconds: {diff_time:.2f} seconds") # 2 decimal places for seconds
    print(f"Total Time in minutes: {diff_time / 60:.2f} minutes") # 60 seconds ≈ 1 minute (used to estimate 1 minute)
    print(f"Total Time in hours: {diff_time / 3600:.2f} hours") # 3600 seconds ≈ 1 hour (used to estimate 1 hour)
    print(f"Total Time in days: {diff_time / 86400:.2f} days") # 86400 seconds ≈ 24 hours (used to estimate 1 day)
    print(f"Total Time in weeks: {diff_time / 604800:.2f} weeks") # 604800 seconds ≈ 7 days (used to estimate 1 week)
    print(f"Total Time in months (approx.): {diff_time / 2592000:.2f} months") # 2592000 seconds ≈ 30 days (used to estimate 1 month)
    print(f"Total Time in years (approx.): {diff_time / 31536000:.2f} years") # 31536000 seconds ≈ 365 days (used to estimate 1 year)
except ValueError:
    print("Invalid input. Please enter a valid integer to end the loop.")
```
