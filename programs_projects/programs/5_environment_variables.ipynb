# How to access environment variables in Python

Environment variables are like secret settings or global values your computer or server keeps. 

They can store information like: Your username/ Database password/ API keys/ System paths



```python
# os.environ from os module allows you to read and modify environment variables in your Python scripts.
import os

# os.environ.get("VARIABLE_NAME") # returns the value of the environment variable named "VARIABLE_NAME".
# os.environ['VARIABLE_NAME'] = 'value' # sets the environment variable named "VARIABLE_NAME" to 'value'.

# os.environ['DB_USER'] = ''
# db_user = os.environ.get('DB_USER')  # Returns None if variable doesn't exist
# db_user = os.environ['DB_USER']  # Raises KeyError if variable doesn't exist
# print(db_user)

print(f'HOME: {os.environ.get("HOME")}') # Prints the HOME environment variable
print(f'USER: {os.environ.get("USER")}') # Prints the USER environment variable
print(f'USERNAME: {os.environ.get("USERNAME")}') # Prints the USERNAME environment variable
print(f'PWD: {os.environ.get("PWD")}') # Prints the PWD environment variable
print(f'LANG: {os.environ.get("LANG")}') # Prints the LANG environment variable
print(f'LANGUAGE: {os.environ.get("LANGUAGE")}') # Prints the LANGUAGE environment variable
print(f'SHELL: {os.environ.get("SHELL")}') # Prints the SHELL environment variable

```
