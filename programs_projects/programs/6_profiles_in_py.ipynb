# How to do a profile a Python script

- Profiling means analyzing our code to see:

    - Which parts take the most time to run

    - How often each function is called

    - How much memory is used (optional, more advanced)

- It helps us understand where our code is slow, and where we can optimize it.

```python
# Use %time — Single run timing — This shows how long that line took to run once.
%time sum([i for i in range(1000000)])
```

```python
# Use %timeit — Multiple run timing —    This runs the line multiple times and gives you an average time.
%timeit sum([i for i in range(1000000)])
```

> Use %%time — Cell timing — This shows how long the entire cell took to run.

```python
%%time
total = 0
for i in range(1000000):
    total += i
```

> Use %%timeit — Cell multiple run timing — This runs the entire cell multiple times and gives you an average time.

```python
%%timeit
total = 0
for i in range(1000000):
    total += i
```

```python
# Use cProfile for profiling the performance of a Python script
import cProfile

def slow_function():
    import time
    time.sleep(2)

def fast_function():
    import time
    time.sleep(0.5)

def main():
    slow_function()
    fast_function()

cProfile.run("main()")
```
