# How to List all Files of a Directory in Python

#### ✅ Approach 1: Using os.listdir()

os.listdir(path) gives you all files and folders in the directory as a list of names (strings). You need to filter out only the files.

```python
import os

# Step 1: Define your directory path
path = '/home/sandeepkumarmishra/Downloads/'

# Step 2: Get all items in that path
all_items = os.listdir(path)

# Step 3: Set counters for files and folders
file_count = 0
folder_count = 0
pdf_file_count = 0
image_file_count = 0

# Step 4: Loop through all items
for item in all_items:
    full_path = os.path.join(path, item)
    
    if os.path.isfile(full_path):
        file_count += 1
        
        # Check for PDF files
        if item.lower().endswith('.pdf'):
            print(f"📄 PDF File: {item}")
            pdf_file_count += 1
        
        # Check for image files (png, jpg, jpeg)
        elif item.lower().endswith(('.png', '.jpg', '.jpeg')):
            print(f"🖼️ Image File: {item}")
            image_file_count += 1

    else:
        folder_count += 1
        print(f"📁 Directory: {item}")

# Step 5: Print summary
print("\n📊 Summary:")
print(f"📄 Total files: {file_count}")
print(f"📑 Total PDF files: {pdf_file_count}")
print(f"🖼️ Total Image files: {image_file_count}")
print(f"📁 Total folders: {folder_count}")
```

> Simple File Explorer using "os" module

```python
import os

# 📥 Step 1: Ask the user for the folder path
folder_path = input("📁 Enter the folder path: ").strip()

# 🚫 Step 2: Check if the path exists and is a directory
if not os.path.exists(folder_path) or not os.path.isdir(folder_path):
    print("❌ The folder path doesn't exist or is not a directory. Please check and try again.")
    exit()

# 🔐 Step 3: Try to list folder contents (handle permission errors)
try:
    all_items = os.listdir(folder_path)
except PermissionError:
    print("🚫 You do not have permission to access this folder.")
    exit()

# 🧮 Step 4: Counters to track overall items
file_count = 0
folder_count = 0

# ✅ Step 4.1: Use dictionary to track counts of specific file types
file_type_counts = {
    'pdf': 0,
    'image': 0,
    'diff': 0,
    'sql': 0,
    'php': 0,
    'text': 0,
    'xlsx': 0,
    'zip': 0,
    'other': 0
}

# 🔁 Step 5: Loop through all items in the folder
for item in all_items:
    full_path = os.path.join(folder_path, item)  # Create full path for file/folder

    # If it's a file
    if os.path.isfile(full_path):
        file_count += 1
        filename = item.lower()  # Convert to lowercase for case-insensitive checks

        # 🧠 Match file types and count using dictionary
        if filename.endswith('.pdf'):
            print(f"📄 PDF File       - {item}")
            file_type_counts['pdf'] += 1
        elif filename.endswith(('.png', '.jpg', '.jpeg')):
            print(f"🖼️ Image File     - {item}")
            file_type_counts['image'] += 1
        elif filename.endswith('.diff'):
            print(f"📝 Diff File      - {item}")
            file_type_counts['diff'] += 1
        elif filename.endswith('.sql'):
            print(f"🗃️ SQL File       - {item}")
            file_type_counts['sql'] += 1
        elif filename.endswith('.php'):
            print(f"🐘 PHP File       - {item}")
            file_type_counts['php'] += 1
        elif filename.endswith('.txt'):
            print(f"📘 Text File      - {item}")
            file_type_counts['text'] += 1
        elif filename.endswith(('.csv', '.xlsx', '.xls')):
            print(f"📊 Excel File     - {item}")
            file_type_counts['xlsx'] += 1
        elif filename.endswith('.zip'):
            print(f"🗜️ ZIP File       - {item}")
            file_type_counts['zip'] += 1
        else:
            print(f"📦 Other File     - {item}")
            file_type_counts['other'] += 1

    # If it's a folder
    elif os.path.isdir(full_path):
        folder_count += 1
        print(f"📁 Directory      - {item}")

# 📊 Step 6: Print summary using the dictionary
print("\n📊 Summary Report")
print("="*50)
print(f"📁 Total folders             : {folder_count}")
print(f"📄 Total files               : {file_count}")
print(f"   - 📑 PDF files            : {file_type_counts['pdf']}")
print(f"   - 🖼️ Image files          : {file_type_counts['image']}")
print(f"   - 📝 Diff files           : {file_type_counts['diff']}")
print(f"   - 🗃️ SQL files            : {file_type_counts['sql']}")
print(f"   - 🐘 PHP files            : {file_type_counts['php']}")
print(f"   - 📘 Text files           : {file_type_counts['text']}")
print(f"   - 📊 Excel files          : {file_type_counts['xlsx']}")
print(f"   - 🗜️ ZIP files            : {file_type_counts['zip']}")
print(f"   - 📦 Other files          : {file_type_counts['other']}")
print("="*50)
```

#### ✅ Approach 2: Using os.scandir()


🧠 Why use it?

It scans the directory and returns DirEntry objects, which are like smart file references.

You can quickly check if the item is a file, folder, or even a symlink without needing to build full paths or make extra function calls.

It’s also faster than os.listdir() for large directories.

```python
import os  # Import the OS module to interact with the operating system

# Define the path where you want to scan files and folders
path = '/home/sandeepkumarmishra/Downloads/'

# Use os.scandir() to efficiently scan the directory
with os.scandir(path) as all_items:
    # Loop through each item in the directory
    for items in all_items:

        # Check if the item is a folder (directory)
        if items.is_dir():
            # Get folder size in bytes and convert to kilobytes (rounded to 2 decimal places)
            size_kb = round(items.stat().st_size / 1024, 2)

            # Print folder name, size, and full path
            print(f"\n📁 Folder Name: {items.name} - {items.stat().st_size} bytes - {size_kb} KB")
            print(f"[📂 Path: {items.path}]")

            # Try to list contents of the folder (some folders might be restricted)
            try:
                sub_folder = os.listdir(items.path)  # List all items inside this folder

                # Print header before listing contents
                print(f"\n🗂️ Contents of folder: {items.name}")
                print("=" * 50)

                # Loop through items inside the subfolder and print their names
                for sub_item in sub_folder:
                    print(f"📄 File: {sub_item}")

                # Print footer line
                print("=" * 50)

            except PermissionError:
                # Handle the case where you don’t have permission to open the folder
                print("❌ Cannot access this folder (Permission Denied)")

        # If the item is a regular file (not a folder)
        elif items.is_file():
            # Get file size in kilobytes (rounded)
            size_kb = round(items.stat().st_size / 1024, 2)

            # Print file name, size, and path
            # print(f"📄 File Name: {items.name} - {items.stat().st_size} bytes - {size_kb} KB")
            # print(f"[📄 Path: {items.path}]\n")
```

#### ✅ Approach 3: Using glob.glob()

- glob stands for "global pattern matching". It works like file search in Windows (*.txt, data_*.csv, etc.).



📌 Things to Remember
- glob.glob() returns file names as strings, not Path objects.
- It won’t tell you if it’s a file or folder—you need to use os.path or pathlib if you need that.
- It’s great for filtering, but not as powerful for deep inspection (e.g., checking size, modified time).


```python
import glob  # Import the glob module for pattern-based file searching

# Set the path to the directory you want to search
path = '/home/sandeepkumarmishra/Downloads/'

# matched_files = glob.glob(path + "/*.txt")  # ✅ List all .txt files in current folder (non-recursive)
matched_files = glob.glob(path + "**/*.txt", recursive=True)  # ✅ Recursively find all .txt files in subfolders

# Loop through the matched file list and print each file path
for file in matched_files:
    print(file)  # 🖨️ Print full path of each matched .txt file
```

#### ✅ Approach 4: Using pathlib.Path.iterdir()

- Why use pathlib?
    - It's object-oriented and cleaner than os-based methods.
    - Works consistently across all operating systems.
    - Lets you use intuitive methods like .is_file(), .suffix, .stat(), etc.


```python
# ✅ Step 1: Import necessary modules
from pathlib import Path       # For working with files and folders in an object-oriented way
from datetime import datetime  # To convert file modified timestamp into readable format

# ✅ Step 2: Define the directory to scan
directory = Path('/home/sandeepkumarmishra/Downloads/')

# ✅ Step 3: Initialize counters for folders, files, and others
total_folder_count = 0
total_file_count = 0
total_other_count = 0

# ✅ Step 4: Set up a dictionary to track counts of different file types
file_type_counts = {
    'pdf': 0,
    'image': 0,
    'diff': 0,
    'sql': 0,
    'php': 0,
    'text': 0,
    'xlsx': 0,
    'zip': 0,
    'other': 0  # for all unknown or unclassified file types
}

# ✅ Step 5: Iterate over each item in the directory
for item in directory.iterdir():

    # 🔍 Check if the current item is a file
    if item.is_file():
        total_file_count += 1  # Increment total file counter

        # Get file size in KB (Bytes ➝ KB)
        size_kb = round(item.stat().st_size / 1024, 2)

        # Get last modified date in human-readable format
        last_modified = datetime.fromtimestamp(item.stat().st_mtime)

        # ✅ Step 6: Classify the file based on its extension
        if item.suffix.lower() == '.txt':
            file_type_counts['text'] += 1
            # print(f"Text File - {item.name} | {size_kb} KB | Modified on: {last_modified}")

        elif item.suffix.lower() in ['.csv', '.xlsx', '.xls']:
            file_type_counts['xlsx'] += 1

        elif item.suffix.lower() in ['.jpeg', '.jpg', '.png']:
            file_type_counts['image'] += 1

        elif item.suffix.lower() == '.pdf':
            file_type_counts['pdf'] += 1

        elif item.suffix.lower() == '.sql':
            file_type_counts['sql'] += 1

        elif item.suffix.lower() == '.php':
            file_type_counts['php'] += 1

        elif item.suffix.lower() == '.zip':
            file_type_counts['zip'] += 1

        elif item.suffix.lower() == '.diff':
            file_type_counts['diff'] += 1

        else:
            file_type_counts['other'] += 1  # Anything not matched above

    # 📁 If the item is a folder
    elif item.is_dir():
        total_folder_count += 1

    # 🧩 If it’s not a file or folder (like symlink or device), count as 'other'
    else:
        total_other_count += 1

# ✅ Step 7: Display a clean summary report
print("\n📊 Summary Report")
print("=" * 50)
print(f"📁 Total folders             : {total_folder_count}")
print(f"📄 Total files               : {total_file_count}")
print(f"🔸 Total Others              : {total_other_count}")
print(f"   - 🖼️ Image files          : {file_type_counts['image']}")
print(f"   - 📘 Text files           : {file_type_counts['text']}")
print(f"   - 📊 Excel files          : {file_type_counts['xlsx']}")
print(f"   - 📄 PDF files            : {file_type_counts['pdf']}")
print(f"   - 🗃️ ZIP files            : {file_type_counts['zip']}")
print(f"   - 💾 SQL files            : {file_type_counts['sql']}")
print(f"   - 🐘 PHP files            : {file_type_counts['php']}")
print(f"   - 🔧 DIFF files           : {file_type_counts['diff']}")
print(f"   - ❓ Other files          : {file_type_counts['other']}")
print("=" * 50)


```
