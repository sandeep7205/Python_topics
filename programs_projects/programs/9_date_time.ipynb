# How to Print Current Date and Time in Python

```python
from datetime import datetime
current_time = datetime.now()
print(f"Currently the TIME is {current_time}")

print(f"Full Date -  {current_time.date()}")
print(f"Year -  {current_time.year}")
print(f"Month -  {current_time.month}")
print(f"Date -  {current_time.day}")

print(f"Full Time -  {current_time.time()}")
print(f"Hour -  {current_time.hour}")
print(f"Minute -  {current_time.minute}")
print(f"Second -  {current_time.second}")
print(f"Micro Second -  {current_time.microsecond}")



formatted_time_12 = current_time.strftime("%m/%d/%y %I:%M %p")
print(f"12hr Formatted Currently the TIME is {formatted_time_12}")
formatted_time_24 = current_time.strftime("%m/%d/%y %H:%M")
print(f"24hr Formatted Currently the TIME is {formatted_time_24}")

week_day = current_time.strftime("%A")
print(f"Weekday - {week_day}")
month = current_time.strftime("%B")
print(f"Month - {month}")


formatted_time_day = current_time.strftime("%A, %d %B %Y At %I:%M %p")
print(f"Today is {formatted_time_day}")


#Show day of the year (e.g., “It’s the 158th day of the year”)
day_of_year = current_time.strftime("%j")
print(f"It’s the {day_of_year}th day of the year.")
```

#### 🔥 Challenge: "How Many Days Left This Year?"
- Write a Python program that tells the user how many days are left until the end of the year.

```python
from datetime import date, datetime

flag = 1  # Try 1 for date.today(), 0 for datetime.now().date()

if flag == 1:
    current_date = date.today()
else:
    current_timestamp = datetime.now()
    current_date = current_timestamp.date()

# Create the end-of-year date (same year as today)
year_end_date = date(current_date.year, 12, 31)

# Show today's date and the year-end date
print(f"Today is: {current_date}")
print(f"Year-end date is: {year_end_date}")

# Calculate how many days are left
remaining_days = year_end_date - current_date
print(f"Days left this year: {remaining_days.days} days")
```

#### 🕰 Mini Project: Terminal-Based Digital Clock

```python
from datetime import datetime
import time
import os

try:
    while True:
        current_timestamp = datetime.now()
        os.system("clear")  # use "cls" on Windows
        # print(current_timestamp.strftime("%A, %d %B"))
        print(current_timestamp.strftime("%H:%M:%S  |or|  %I:%M:%S %p"))
        time.sleep(1)
except KeyboardInterrupt:
    print("\nClock stopped by user. Goodbye!")
```
