# How to Print Current Date and Time in Python

```python
from datetime import datetime
current_time = datetime.now()
print(f"Currently the TIME is {current_time}")

print(f"Full Date -  {current_time.date()}")
print(f"Year -  {current_time.year}")
print(f"Month -  {current_time.month}")
print(f"Date -  {current_time.day}")

print(f"Full Time -  {current_time.time()}")
print(f"Hour -  {current_time.hour}")
print(f"Minute -  {current_time.minute}")
print(f"Second -  {current_time.second}")
print(f"Micro Second -  {current_time.microsecond}")



formatted_time_12 = current_time.strftime("%m/%d/%y %I:%M %p")
print(f"12hr Formatted Currently the TIME is {formatted_time_12}")
formatted_time_24 = current_time.strftime("%m/%d/%y %H:%M")
print(f"24hr Formatted Currently the TIME is {formatted_time_24}")

week_day = current_time.strftime("%A")
print(f"Weekday - {week_day}")
month = current_time.strftime("%B")
print(f"Month - {month}")


formatted_time_day = current_time.strftime("%A, %d %B %Y At %I:%M %p")
print(f"Today is {formatted_time_day}")


#Show day of the year (e.g., “It’s the 158th day of the year”)
day_of_year = current_time.strftime("%j")
print(f"It’s the {day_of_year}th day of the year.")
```

#### 🔥 Challenge: "How Many Days Left This Year?"
- Write a Python program that tells the user how many days are left until the end of the year.

```python
from datetime import date, datetime

flag = 1  # Try 1 for date.today(), 0 for datetime.now().date()

if flag == 1:
    current_date = date.today()
else:
    current_timestamp = datetime.now()
    current_date = current_timestamp.date()

# Create the end-of-year date (same year as today)
year_end_date = date(current_date.year, 12, 31)

# Show today's date and the year-end date
print(f"Today is: {current_date}")
print(f"Year-end date is: {year_end_date}")

# Calculate how many days are left
remaining_days = year_end_date - current_date
print(f"Days left this year: {remaining_days.days} days")
```

#### 🕰 erminal-Based Digital Clock

```python
from datetime import datetime
import time
import os

try:
    while True:
        current_timestamp = datetime.now()
        os.system("clear")  # use "cls" on Windows
        # print(current_timestamp.strftime("%A, %d %B %Y"))
        print(current_timestamp.strftime("%H:%M:%S  |or|  %I:%M:%S %p"))
        time.sleep(1)
except KeyboardInterrupt:
    print("\nClock stopped by user. Goodbye!")
```

#### 🎂 Birthday Countdown

```python
from datetime import date
import calendar

# 🔁 Function to get and validate user input (for year, month, or day)
# Prompts user with input(), checks range, and handles quit option
def get_valid_input(field_type):
    while True:
        input_raw = input(f"Enter your birth {field_type.capitalize()} (or 'q' to quit): ").strip()

        # 🛑 Allow user to quit anytime by typing 'q'
        if input_raw.lower() == 'q':
            print("👋 Exiting program. See you next time!")
            return None  # Ends the current input process

        # ✅ Try converting the input to an integer (for validation)
        try:
            input_value = int(input_raw)
        except ValueError:
            print("⚠️ Please enter a valid number or 'q' to quit.")
            continue

        # 🧠 Use match-case to handle validation for each field
        match field_type:
            case "date":
                if 1 <= input_value <= 31:
                    return input_value
                else:
                    print("❌ Date must be between 1 and 31.")
            case "month":
                if 1 <= input_value <= 12:
                    return input_value
                else:
                    print("❌ Month must be between 1 and 12.")
            case "year":
                if input_value > 0 and len(str(input_value)) == 4:
                    return input_value
                else:
                    print("❌ Year must be a 4-digit positive number.")
            case _:
                print("❌ Unknown field type. This shouldn't happen.")

# ✨ Function to convert a day number (like 1) into an ordinal (like 1st, 2nd, 3rd)
# Used to make the birthday message look natural
def get_ordinal_suffix(day):
    if 10 <= day % 100 <= 20:  # Covers special cases like 11th, 12th, 13th
        suffix = 'th'
    else:
        suffix = {1: 'st', 2: 'nd', 3: 'rd'}.get(day % 10, 'th')
    return str(day) + suffix

# 🎯 Main logic for the birthday tracker — wrapped in a function for reusability
def run_birthday_tracker():
    try:
        # 🧩 Step 1: Ask for birth year, month, and day with validation
        birth_year = get_valid_input('year')
        if birth_year is None: return

        birth_month = get_valid_input('month')
        if birth_month is None: return

        birth_day = get_valid_input('date')
        if birth_day is None: return

        # 📅 Step 2: Check if the given date is valid for the specified month/year
        # e.g., Feb 30 is invalid
        max_day_in_month = calendar.monthrange(birth_year, birth_month)[1]
        if birth_day > max_day_in_month:
            print("❌ That day doesn't exist in the selected month and year.")
            return

        # ✅ Create a proper birthdate object
        birth_date = date(birth_year, birth_month, birth_day)
        print(f"\n📅 Your birth date is: {birth_date}")

        # 🕒 Step 3: Get today's date and calculate the birthday this year
        today = date.today()
        this_year = today.year
        upcoming_birthday_date = date(this_year, birth_month, birth_day)

        # 🧮 Step 4: Calculate how many days left until next birthday
        if today <= upcoming_birthday_date:
            days_until_birthday = (upcoming_birthday_date - today).days
        else:
            # 📆 If birthday has passed, prepare for next year's birthday
            next_year = this_year + 1
            upcoming_birthday_date = date(next_year, birth_month, birth_day)
            days_until_birthday = (upcoming_birthday_date - today).days

        # 🚫 Step 5: If birth date is in the future, show error
        if birth_date > today:
            print("❌ Birth date cannot be in the future.")
            return

        # 🎂 Step 6: Calculate current age
        has_birthday_occurred = (today.month, today.day) >= (birth_month, birth_day)
        calculated_age = this_year - birth_year - (0 if has_birthday_occurred else 1)
        print(f"🎈 You are currently {calculated_age} years old.")

        # 🎨 Step 7: Prepare a nicely formatted birthday string
        formatted_year = upcoming_birthday_date.year                         # e.g., 2025
        formatted_month = upcoming_birthday_date.strftime("%B")             # e.g., November
        formatted_weekday = upcoming_birthday_date.strftime("%A")           # e.g., Tuesday
        formatted_day = get_ordinal_suffix(upcoming_birthday_date.day)      # e.g., 27th

        birthday_string = f"{formatted_weekday}, {formatted_day} {formatted_month}, {formatted_year}"

        # 📢 Step 8: Display birthday countdown message
        if days_until_birthday == 0:
            print("🎉 Happy Birthday to You! 🎂")
        else:
            print(f"⏳ {days_until_birthday} days left until your next birthday on {birthday_string}")

    except KeyboardInterrupt:
        # 🙋 Graceful exit if user hits Ctrl+C
        print("\n👋 Program interrupted. Goodbye!")

# ▶️ Entry point: Run the tracker
run_birthday_tracker()
```
