# Data Types
#### Built-in Data Types
```
Text Type:	str
Numeric Types:	int, float, complex
Sequence Types:	list, tuple, range
Mapping Type:	dict
Set Types:	set, frozenset
Boolean Type:	bool
Binary Types:	bytes, bytearray, memoryview
None Type:	NoneType
```
#### Setting the Data Type



````python
x = "Hello World"	#str	
print("Value:", x, " →→→ Type:", type(x))

x = 20	#int	
print("Value:", x, " →→→ Type:", type(x))

x = 20.5	#float	
print("Value:", x, " →→→ Type:", type(x))

x = 1j	#complex	
print("Value:", x, " →→→ Type:", type(x))

x = ["apple", "banana", "cherry"]	#list	
print("Value:", x, " →→→ Type:", type(x))

x = ("apple", "banana", "cherry")	#tuple	
print("Value:", x, " →→→ Type:", type(x))

x = range(6)	#range	
print("Value:", x, " →→→ Type:", type(x))

x = {"name" : "John", "age" : 36}	#dict	
print("Value:", x, " →→→ Type:", type(x))

x = {"apple", "banana", "cherry"}	#set	
print("Value:", x, " →→→ Type:", type(x))

x = frozenset({"apple", "banana", "cherry"})	#frozenset	
print("Value:", x, " →→→ Type:", type(x))

x = True	#bool	
print("Value:", x, " →→→ Type:", type(x))

x = b"Hello"	#bytes	
print("Value:", x, " →→→ Type:", type(x))

x = bytearray(5)	#bytearray	
print("Value:", x, " →→→ Type:", type(x))

x = memoryview(bytes(5))	#memoryview	
print("Value:", x, " →→→ Type:", type(x))

x = None	#NoneType
print("Value:", x, " →→→ Type:", type(x))
````

## Numbers
``` Note: We cannot convert complex numbers into another number type.```

````python
# int
x = 1
y = 35656222554887711
z = -3255522
print(type(x), type(y), type(z))

# float
x = 1.10
y = 1.0
z = -35.59
print(type(x), type(y), type(z))

# float
x = 35e3
y = 12E4
z = -87.7e100
print(type(x), type(y), type(z))

# complex
x = 3+5j
y = 5j
z = -5j
print(type(x), type(y), type(z))
````

#### Slicing Strings

````python
b = "Hello, World!"
print(b[2:5]) # start point index - 2 &  end point index - 5  
print(b[:5]) # start point index - 0 &  end point index - 5  
print(b[2:]) # start point index - 2 &  end point index - end
print(b[-5:-2]) # From: "o" in "World!" (position -5) - To, but not included: "d" in "World!" (position -2):
````

#### Format - Strings [f-strings]

````python
price = 36
txt = f"My name is John, I am {price}"    #A placeholder can contain variables, operations, functions, and modifiers to format the value.
print(txt)
txt2 = f"The price is {price:.2f} dollars"  #A modifier is included by adding a colon : followed by a legal formatting type, like .2f which means fixed point number with 2 decimals
print(txt2)
txt3 = f"The price is {price * 3} dollars" 
print(txt3)
````

#### Escape Characters

````python
# Single Quote	
txt = 'It\'s alright.'
print(txt) 

# Double Quote	
txt = "We are the so-called \"Vikings\" from the north."
print(txt)

# Backslash
txt = "This will insert one \\ (backslash)."
print(txt) 

# New Line	
txt = "This will be a \n new line."
print(txt) 

# Carriage Return	
txt = "Hello\rWorld!"
print(txt) 

# Tab
txt = "Hello\tWorld!"
print(txt) 

# This example erases one character (backspace):
txt = "Hello \bWorld!"
print(txt) 

# Form Feed	
txt = "Hello \fWorld11!"
print(txt) 

# Octal value - A backslash followed by three integers will result in a octal value:
txt = "\110\145\154\154\157"
print(txt) 

# Hex value	- A backslash followed by an 'x' and a hex number represents a hex value:
txt = "\x48\x65\x6c\x6c\x6f"
print(txt) 
````

#### Arithmetic Operators
 - Arithmetic operators are used with numeric values to perform common mathematical operations

````python
x = 15
y = 4
# Arithmetic operators
print(f'Addition: {x} + {y} = {x + y}')
print(f'Subtraction: {x} - {y} = {x - y}')
print(f'Multiplication: {x} * {y} = {x * y}')
print(f'Division: {x} / {y} = {x / y}') # (returns a float)
print(f'Modulus: {x} % {y} = {x % y}')
print(f'Exponentiation: {x} ** {y} = {x ** y}')
print(f'Floor division: {x} // {y} = {x // y}') # (returns an largest integer less then result)
````

#### Assignment Operators
 - Assignment operators are used to assign values to variables:

````python
# =  (Assign)
x = 5
print("=  →", x)

# +=  (Add and Assign)
x = 5
x += 3   # same as x = x + 3
print("+= →", x)

# -=  (Subtract and Assign)
x = 5
x -= 3   # same as x = x - 3
print("-= →", x)

# *=  (Multiply and Assign)
x = 5
x *= 3   # same as x = x * 3
print("*= →", x)

# /=  (Divide and Assign)
x = 9
x /= 3   # same as x = x / 3
print("/= →", x)

# %=  (Modulus and Assign)
x = 10
x %= 3   # same as x = x % 3
print("%= →", x)

# //=  (Floor Divide and Assign)
x = 10
x //= 3   # same as x = x // 3
print("//= →", x)

# **=  (Exponent and Assign)
x = 2
x **= 3   # same as x = x ** 3
print("**= →", x)

# &=  (Bitwise AND and Assign)
x = 5
x &= 3    # same as x = x & 3
print("&= →", x)

# |=  (Bitwise OR and Assign)
x = 5
x |= 3    # same as x = x | 3
print("|= →", x)

# ^=  (Bitwise XOR and Assign)
x = 5
x ^= 3    # same as x = x ^ 3
print("^= →", x)

# >>=  (Right Shift and Assign)
x = 8
x >>= 2   # same as x = x >> 2
print(">>= →", x)

# <<=  (Left Shift and Assign)
x = 8
x <<= 2   # same as x = x << 2
print("<<= →", x)

# :=  (Walrus Operator)
# Assigns value *inside* expression
print(":= →", (x := 3))   # sets x = 3 and prints it
print("x is now:", x)
````
