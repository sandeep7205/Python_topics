# Lists

- Lists are used to store multiple items in a single variable.
- Lists are created using square brackets: `["apple", True, 2, 3.55, 20]` or `list(("apple", True, 2, 3.55, 20))`
- List items are 
    - indexed (starts with 0)
    - ordered (index order 0, 1, 2, etc)
    - changeable(modifie, add, and remove items)
    - allow duplicate values.


🔍 Think like this…
- A list is like a row of seats in a theater:
    - Each seat has a fixed number (index)
    - You can remove people (items)
    - You can insert new people in the middle
    - You can replace a group of seats with more seats or fewer seats
    - But the seating order (left → right) always matters and remains consistent
    - Only you change the order, Python doesn't shuffle for you.

````python
# Lists allow Duplicate Values.
# A list's Items with any Data Types (strings, integers and boolean values):
thislist = ["apple", "cherry", "apple", True, 2, 3.55, 20j, "3.55", 2, "apple"]
print("Original List → ", thislist)

# List Length
print("len(list) → Number of items in the list → ", len(thislist))

# List type
print("type(list) → Data type of a list → ", type(thislist))

# list() Constructor
thislist = list(("apple", "cherry", "apple", True, 2, 3.55, 20j, "3.55", 2, "apple")) # note the double round-brackets
print("original list() → ", thislist)

# Access List Items - Index id (The first item has index 0)
print("4th index or 5th Item → ", thislist[4])

# Access List Items - Negative Indexing (Indexwith -1 start from the end/Last item )
print("-3rd index or 3rd Item from end → ", thislist[-3])

#Range of Indexes - The search will start at index 2 (included) and end at index 5 (not included).
print("2nd to 4th index (5th excluded) → ", thislist[2:5]) 
print("0th to 3rd index (4th excluded) → ",thislist[:4]) 
print("2nd to ending index → ",thislist[2:])
print("-4th to -2nd index (-1st excluded) → ",thislist[-4:-1]) #index -1 (not included)

# Check if Item Exists
if "apple" in thislist:
  print("Yes, 'apple' is in the fruits list")

````

#### Change List Items

````python
thislist = list(("apple", "banana", "cherry")) # note the double round-brackets
print("original list - ", thislist)

# Change Item Value
thislist[2] = "blackcurrant"
print("2nd index/3rd items got replace with → ", thislist)

# Change a Range of Item Values 
thislist[1:3] = ["blackcurrant", "watermelon"] # Change the values "banana" and "cherry" with the values "blackcurrant" and "watermelon":
    # The length of the list will change when the number of items inserted does not match the number of items replaced.
print("1st and 2nd index got replace with → ",thislist)

thislist[1:2] = ["34", "45", 89] 
    # If you insert less/more items than you replace, the new items will be inserted where you specified, and the remaining items will move accordingly:
print("1st index got replace and other items add into list → ",thislist)

thislist[1:3] = ["watermelon"]
    # If you insert less items than you replace, the new items will be inserted where you specified, and the remaining items will move accordingly:
print("1st index got replace with → ",thislist)

thislist[2:4] = []
print("2nd, 3rd index got replace blank list → ",thislist)
````

#### Add List Items

````python
thislist = list(("apple", "banana", "cherry"))
print("original list - ", thislist)

#Insert Items
thislist.insert(5, 99.99) # insert a list item at a specified index
print("insert(): add item into 5th index of the list → ", thislist)

# Append Items
thislist.append("orange") #add an item to the end of the list
print("append(): add item at the end of the list → ", thislist)

tropical = list(("mango", False)) 
# append elements from another list to the current list and The elements will be added to the end of the list.
thislist.extend(tropical)
print(f"extend(): By Merging list {tropical} → {thislist}")

# Can add any iterable object (tuples, sets, dictionaries etc.) with lists
thistuple = (24e3j, "mango")
thislist.extend(thistuple) #Add elements of a tuple to a list
print(f"extend(): By Merging tuple {thistuple} → {thislist}")
````

````python


thislist.remove("mango") #Remove the first occurance of "mango"
print(thislist)

thislist.pop() #Remove the last item
print(thislist)

thislist.pop(1) #Remove the mentioned index item
print(thislist)

del thislist[0] #Remove the mentioned index item
print(thislist)

# thislist.clear() #Clear the list content
# print(thislist)

# del thislist #Delete the entire list:
# print(thislist)
````

#### For/While Loop Through a List

````python
thislist = ["apple", "banana", "cherry"]

# for Loop
for x in thislist:
  print(x)

for i in range(len(thislist)):
  print(thislist[i])

# While Loop
i = 0
while i < len(thislist):
  print(thislist[i])
  i = i + 1
````

#### Looping Using List Comprehension  
```
[List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list.]
```
> Syntax  -  newlist = [expression for item in iterable if condition == True]

````python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]

#With no if statement
[print(x) for x in fruits] 

#Without list comprehension
newlist1 = []
for x in fruits:
  if "a" in x:
    newlist1.append(x)
print(newlist1)

#With list comprehension
newlist2 = [x for x in fruits if "a" in x]
print(newlist2)

newlist3 = [x if x != "banana" else "orange" for x in fruits] #Return "orange" instead of "banana"
print(newlist3)
````

#### Sort Lists

````python
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]

thislist.sort() #Sort the list alphabetically
print(thislist)

thislist.sort(reverse = True) #Sort the list descending
print(thislist)

thislist.reverse() # Reverse the order of the list items:
print(thislist)

#Sort the list based on how close the number is to 50:
thisnumlist = [100, 50, 65, 82, 23]
def myfunc(n):
  return abs(n - 50)

thisnumlist.sort(key = myfunc)
print(thisnumlist)
````

#### Copy Lists

````python
thislist = ["apple", "banana", "cherry"]
mylist = thislist
print(mylist)
mylist1 = thislist.copy() # Make a copy of a list with the copy() method:
print(mylist1)
mylist2 = list(thislist) #Make a copy of a list with the list() method:
print(mylist2)
````

#### Join Lists

````python
list1 = ["a", "b", "c"]
list2 = [1, 2, 3]
list3 = [True, 2e3j, 3.757]

join_list = list1 + list2 + list3
print(join_list)

#Append list1, list2, list3 into join_list using append():
join_list = []
for x,y,z in zip(list1,list2,list3):
  join_list.append(x)
  join_list.append(y)
  join_list.append(z)

print(join_list)

list1.extend(list2) #Use the extend() method to add list2 at the end of list1:
print(list1)
````
