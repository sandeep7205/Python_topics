## Tuples ( )

Tuples are used to store multiple items in a single variable.

A tuple is a collection that is:

- **Ordered** → items have a fixed position (index-based)
- **Unchangeable (immutable)** → items cannot be modified after creation
- **Allows duplicate values** → same value can appear multiple times

Tuples are written using **round brackets `( )`**

---

### 🧠 Important Beginner Insights

Once a tuple is created, you **cannot**:

- Add items
- Remove items
- Change existing items

Tuples are **faster than lists** because they are immutable.

Tuples are often used for:

- Fixed data (coordinates, RGB values)
- Returning multiple values from a function
- Unpacking values safely

---

Think of a tuple like:

> **A sealed box 📦**  
> You can look inside, but you can’t change what’s inside.


```python
thistuple = ("apple", "banana", "cherry", "apple", "cherry")
print("tuple with duplicate values - ", thistuple)

# # thistuple[0] = 67786  # TypeError: 'tuple' object does not support item assignment

print("[0]: first index of tuple - ", thistuple[0])

#Print the number of items in the tuple
print("len(): length of tuple - ", len(thistuple)) 

thistuple_new = tuple(("apple", "banana", "cherry")) # note the double round-brackets
print("tuple(()) constructor - ", thistuple_new)
```

#### Create Tuple With One Item

- To create a tuple with only one item, you have to add a comma `,` after the item, otherwise Python will not recognize it as a tuple.


```python
thistuple = ("apple",)
print("comma ',' ('apple',): need for single item - ", type(thistuple)) # tuples are defined as objects with the data type 'tuple':

# NOT a single item tuple without ',' 
thistuple = ("apple")
print("not a tuple without comma ',' ('apple'): variable type is - ",type(thistuple))
```

### Access Tuple

#### Access Tuple Items

```python
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple)

print("second item in the tuple: - ", thistuple[1])
print("Negative Indexing: last item in the tuple - ", thistuple[-1])

print("\n------------\n")

# Range of Indexes:
print("Range of Indexes:[1:5] second to fifth item in the tuple: - ",thistuple[1:5]) # will start at index 2 (included) and end at index 5 (not included).
print("[:5]: returns the items from the beginning to, but NOT included 5th index - ", thistuple[:5])
print("[3:]:returns the items from 3rd index and to the end - ", thistuple[3:])

print("\n------------\n")

# Range of Negative Indexes
print("Negetive Range of Indexes: from index -7 (included) to index -3 (excluded) [-7:-3] - ", thistuple[-7:-3])
```

#### Check if Item Exists
- To determine if a specified item is present in a tuple use the in keyword:

```python
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple)

if "cherry" in thistuple:
    print("cherry exists in tuple")
```

### Update Tuples
- Tuples are `unchangeable, or immutable`    
    - Tuples are unchangeable, meaning that you cannot change, add, or remove items once the tuple is created with some workarounds.
    - But there is a workaround. You can convert the `tuple into a list, change the list, and convert the list back into a tuple`.

#### Change Tuple Values

```python
# Convert the tuple into a list to be able to change the value in it:
x = ("apple", "banana", "cherry")
print("original Tuple - ", x)
#convert into a list:
y = list(x) 
print("Convert into a list() - ", y)
#edit the list:
y[1] = "kiwi"
print("add/replace item using index in list mode - ", y)
#convert it back into a tuple
x = tuple(y) 
print("Tuple after add/edit items - ", x)

print("\n------------\n")

y.append("orange") 
print("append(): value add to the converted list - ", y)

z = ("melon",) #Create a new tuple with the value "melon",
print("create a new tuple having single item with ',' - ",  z)

#add new tuple with old list (converted from tuple):
y += z 
y.extend(("mango",)) 
print("Add new tuple with list", y)

#remove "apple" from list
y.remove("apple") 
print("Remove th eitem from list", y)

print("Tuple before list convertion - ", x)
x = tuple(y)
print("Tuple after convertion of list  - ", x)

#The del keyword can delete the tuple completely:
del x
print("tuple deleted")
```

#### Unpack Tuples

- When we create a tuple, we normally assign values to it. This is called `packing` a tuple: 
`fruits = ("apple", "banana", "cherry")`

- But, in Python, we are also allowed to extract the values back into variables. This is called `unpacking`: 
`(green, yellow, red) = fruits`

> Note: The number of variables must match the number of values in the tuple, if not, you must use an asterisk to collect the remaining values as a list.


- Using Asterisk `*`
    - If the number of variables is less than the number of values, you can add an * to the variable name and the values will be assigned to the variable as a list:
    - If the asterisk is added to another variable name than the last, Python will assign values to the variable until the number of values left matches the number of variables left.
    - ` multiple starred '*' expressions in assignment is not allowed`
    - `_ means “ignore this value”`

```python
fruits = ("apple", "banana", "cherry")
print("Packing: Assign tuple values to a variable - ", fruits)

#Unpacking a tuple:
(green, yellow, red) = fruits
print("UnPacking: Index 0 / Value 1 - ", green)
print("UnPacking: Index 1 / Value 2 - ", yellow)
print("UnPacking: Index 2 / Value 3 - ", red)

print("\n------------\n")

#Using Asterisk[*] Assign the rest of the values as a list called "red":
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")
print("Tuple - ", fruits)
print("\n--add * to last variable--\n")
(green, yellow, *red) = fruits
print("UnPacking: Index 0 / Value 1 - ", green)
print("UnPacking: Index 1 / Value 2 - ", yellow)
print("UnPacking '*': Index 2 / Value 3 to last index/value as list - ", red)
print("\n--add * before last variable--\n")
(green, *yellow, red) = fruits
print("UnPacking: Index 0 / Value 1 - ", green)
print("UnPacking '*': Index 1 / Value 2 to last index 3 / value 4 as list - ", yellow)
print("UnPacking: Index 4 / Value 5  - ", red)

print("\n------------\n")

# _ means “ignore this value”
numbers = (1, 2, 4.66678444, 6, 3.33, 6, 1, 4, 8.5, 10, 4)
print(numbers)
# # (n1, n2, n3, *n4, n5, *n6) = numbers # SyntaxError: multiple starred expressions in assignment
(n1, n2, n3, _, *n5, n6, _) = numbers  
print(n1)
print(n2)
print(n3)
print(n5)
print(n6)
```

#### Loop Tuples
- You can loop through the `tuple items` by using a `for & while loop`.

```python
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")
print("Tuple - ", fruits)

print("\n--Iterate through the items and print the values using for loop--\n")
for x in fruits:
  print(x)

print("\n--Print all items by referring to their index number: with range(len(Tuple)) --\n")

print("len(): Tuple length - ", len(fruits))
print("range(0, len(tuple)): Range of Tuple length - ", range(len(fruits))) # range(start [default 0], stop, step)

for i in range(len(fruits)):
  print(fruits[i])

print("\n--Print all items, using a while loop to go through all the index numbers: with len(tuple)--\n")

i = 0
while i < len(fruits):
  print(f"index {i} value is {fruits[i]}")
  i += 1
```

#### Join Tuples

- To join two or more tuples you can use the `+` operator:
- To multiply the content of a tuple a given number of times, you can use the `*` operator:

```python
#Join two tuples:
tuple1 = ("a", "b786" , 3j+8)
tuple2 = (1, "a", 3.89)
tuple3 = tuple1 + tuple2 
print("Join two tuples: using '+' - ", tuple3)

# Multiply the fruits tuple by 2:
fruits = ("apple", "banana", "cherry")
mytuple = fruits * 2
print("Multiply the fruits tuple by 2:using '*' - ", mytuple)
```

#### 📋 Python Tuple Methods

| Method        | Description                                                                 |
|--------------|-----------------------------------------------------------------------------|
| `count()`   | Returns the number of times a specified value occurs in a tuple                                       |
| `index()`    | Searches the tuple for a specified value and returns the position of where it was found |

```python
tuple1 = ("a", "b", "786" , 3j+8)
tuple2 = (1, "a", 3.89)
tuple3 = tuple1 + tuple2 
print("tuples - ", tuple3)
print("count('a'): Tuple item occured - ", tuple3.count('a'))
print("index('a'): Tuple item first occurance index - ", tuple3.index('a'))
```
