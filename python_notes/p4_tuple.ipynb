## Tuples

- Tuples are used to store multiple items in a single variable.
- A tuple is a collection which is `ordered, unchangeable, and allow duplicate values`. 
- Tuples are written with round `()` brackets.

````python
thistuple = ("apple", "banana", "cherry", "apple", "cherry")
print("tuple with duplicate values - ", thistuple)

print("[0]: first index of tuple - ", thistuple[0])

#Print the number of items in the tuple
print("len(): length of tuple - ", len(thistuple)) 

thistuple_new = tuple(("apple", "banana", "cherry")) # note the double round-brackets
print("tuple(()) constructor - ", thistuple_new)
````

#### Create Tuple With One Item

- To create a tuple with only one item, you have to add a comma `,` after the item, otherwise Python will not recognize it as a tuple.


````python
thistuple = ("apple",)
print("comma ',' ('apple',): need for single item - ", type(thistuple)) # tuples are defined as objects with the data type 'tuple':

#NOT a tuple
thistuple = ("apple")
print("without comma ',' ('apple'): not a tuple - ",type(thistuple))
````

#### Tuple Items - Data Types
- Tuple items can be of any data type:

````python
mytuple = (90, 45.67, 3j+6, "680hkh", True)
print("tuple contains different data types - ", mytuple)
````

### Access Tuple

#### Access Tuple Items

````python
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple)

print("second item in the tuple: - ", thistuple[1])

print("Negative Indexing: last item in the tuple - ", thistuple[-1])

# Range of Indexes:
print("Range of Indexes:[1:5] second to fifth item in the tuple: - ",thistuple[1:5]) # will start at index 2 (included) and end at index 5 (not included).
print("[:5]: returns the items from the beginning to, but NOT included 5th index - ", thistuple[:5])
print("[3:]:returns the items from 3rd index and to the end - ", thistuple[3:])

# Range of Negative Indexes
print("Negetive Range of Indexes: from index -7 (included) to index -3 (excluded) [-7:-3] - ", thistuple[-7:-3])
````

#### Check if Item Exists
- To determine if a specified item is present in a tuple use the in keyword:

````python
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple)

if "cherry" in thistuple:
    print("cherry exists in tuple")
````

### Update Tuples
- Tuples are unchangeable, meaning that you cannot change, add, or remove items once the tuple is created with some workarounds.

#### Change Tuple Values

````python
# Convert the tuple into a list to be able to change the value in it:
x = ("apple", "banana", "cherry")
y = list(x) #convert into a list:
y[1] = "kiwi"
y.append("orange") #add "orange" in list mode
z = ("melon",) #Create a new tuple with the value "melon",
y += z #add that tuple with old tuple:
y.remove("apple") #remove "apple" from list
print(y)
x = tuple(y) #convert it back into a tuple

#The del keyword can delete the tuple completely:
# del x   
print(x)
````

#### Unpack Tuples

> When we create a tuple, we normally assign values to it. This is called "packing" a tuple: 

```fruits = ("apple", "banana", "cherry")```

> But, in Python, we are also allowed to extract the values back into variables. This is called "unpacking": 

```(green, yellow, red) = fruits```

````python
#Unpacking a tuple:
fruits = ("apple", "banana", "cherry")
(green, yellow, red) = fruits

print(green)
print(yellow)
print(red)

#Using Asterisk[*] Assign the rest of the values as a list called "red":
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")

(green, yellow, *red) = fruits

# print(green)
# print(yellow)
# print(red)

(green, *tropic, red) = fruits

print(green)
print(tropic)
print(red)
````

#### Loop Tuples

````python
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")
for x in fruits:
  print(x)

for i in range(len(fruits)):
  print(fruits[i])

while i < len(fruits):
  print(fruits[i])
  i = i + 1
````

#### Join Tuples

````python
#Join two tuples:
tuple1 = ("a", "b" , "c")
tuple2 = (1, 2, 3)
tuple3 = tuple1 + tuple2 
print(tuple3)

# Multiply the fruits tuple by 2:
fruits = ("apple", "banana", "cherry")
mytuple = fruits * 2
print(mytuple)
````
