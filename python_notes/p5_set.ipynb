## Sets { }

A set is a collection that is:

- **Unordered**
- **Unindexed**
- **Unchangeable***  
- **Does not allow duplicate values**

> *Unchangeable means you cannot change existing items,  
> but you **can remove items and add new items**.*

---

### 🧠 Unindexed

- You **cannot be sure about the order** in which items appear
- Items **cannot be accessed using index or key**

---

### 🔀 Unordered

- Items in a set do **not** have a defined order
- Items may appear in a **different order every time**
- You **cannot access** set items using an index or key

---

### 🔒 Unchangeable

- Once a set is created, you **cannot change existing items**
- But you **can**:
  - Add new items
  - Remove existing items

---

### 🚫 Duplicates Not Allowed

- Sets **cannot contain duplicate values**
- If duplicates are added, Python automatically removes them

---

### 🔍 Mental Picture

Think of a set like:

> **A bag of unique items 🎒**

### Join Sets

- There are several ways to join two or more sets in Python.

##### 🔗 Union (`|`)

- The `union()` method **joins all items from both sets**.
  - Returns a **new set**
  - Can join **sets with other iterables** (lists, tuples, etc.)
  - **Duplicates are automatically removed**

- **Set Operator:** `|`
  - Works **only with sets**
  - Same result as `union()`

```python
#union()

set1 = {"a", "b", "c"}
set2 = {0, 1, 2e4j, 3.87}
set3 = {"John", True, False}

thislist = [1, 2, 3.45]
thistuple = (True, 3e2j+9, False)
thisstring = "807"

join_set1 = set1.union(set2, set3)
print("union(): join multiple sets and return new set - ", join_set1)
print("--------")

#Use | instead of union() to join two sets & allows only to join sets with sets 
join_set2 = set1 | set2 | set3 
print("operator `|`: join multiple sets and return new set - ",join_set2)
print("--------")

#Join a set with a tuple
join_set3 = join_set1.union(thistuple, thislist, thisstring) 
# #join_set3 = join_set1 | thistuple | thistuple2  # unsupported operand type(s) for |: 'set' and 'tuple'
print("union(): join multiple 'tuple, list, string' at once and return new set - ", join_set3)
print("--------")
```

##### Update()
- The `update()` method also **joins all items from another set**.
  - **Modifies the original set**
  - Does **not** return a new set
  - Duplicates are automatically removed

```python
#update()

set1 = {"a", "b", "c"}
set2 = {0, 1, 2e4j, 3.87}
set3 = {"John", True, False}

thislist = [1, 2, 3.45]
thistuple = (True, 3e2j+9, False)
thisstring = "807"

set1.update(set2)
set1.update(set3)
set1.update(thistuple)
set1.update(thislist)
set1.update(thisstring)
print("update(): join 'set, tuple, list, string' one after another with main set - ", set1)
```

##### 🔁 Intersection (`&`)

- The `intersection()` method:
  - **Keeps ONLY the common items**
    - Values True or 1, False or 0 are same
  - Returns a **new set**

- **Set Operator:** `&`
  - Works **only with sets**
  - Same result as `intersection()`


```python
# #intersection() -  return a new set, by keeping ONLY the duplicates

set1 = {"apple", "banana", "cherry", 1}
set2 = {"google", "microsoft", "apple"}

print("SET-1: ", set1)
print("SET-2: ", set2)
print("--------")

# #Join set1 and set2, but keep only the duplicates:
set3 = set1.intersection(set2) 
print("intersection(): Return a new set with elements common to the set and all others - ", set3)
print("--------")

# # use the & operator instead of the intersection() method, and allows only to join sets with sets.
set3 = set1 & set2
print("operator `&`: Return a new set with elements common to the set and all others - ",set3)
print("--------")

# Join sets that contains the values True, False, 1, and 0, and see what is considered as duplicates:
set4 = {"apple", 1,  "banana", 0, "cherry"}
set5 = {False, "google", 1, "apple", 2, True}
print("SET-4: ", set1)
print("SET-5: ", set2)
print("--------")
set6 = set4.intersection(set5)
print("intersection(): sets items True or 1 and False or 0 consider as same - set6 - ", set6)

```

##### 🔁 `intersection_update()`

- Keeps **ONLY the common items**
- **Modifies the original set**
- Returns `None`
- Equivalent to the `&=` operator

```python
# #The intersection_update() method will also keep ONLY the duplicates, but it will change the original set instead of returning a new set.
set1 = {"apple", 1,  "banana", 0, "cherry"}
set2 = {False, "google", 1, "apple", 2, True}

print("SET-1: ", set1)
print("SET-2: ", set2)
print("--------")

set1.intersection_update(set2)
print("intersection_update(): Update the set, keeping only elements found in it and other set - ", set1)
print("--------")
thislist = ["apple", 1, 2, 3.87]
thistuple = ("apple", True, 3e2j+9, False)
thisstring = "apple"
set1.intersection_update(thislist, thistuple)
print("intersection_update(): Update the set, keeping only elements found in it and all others set, list, tuple - ", set1) 
print("--------")
singleset= {"apple"}
singleset.update(thisstring) 
print(f"SET after update() string {thisstring} - {singleset}")
print("--------")
set1.intersection_update(thisstring)
# Strings are iterable in Python : "apple" → 'a', 'p', 'p', 'l', 'e'
print("intersection_update(): Update the set, keeping only elements found in it and string - ", set1) 
print("--------\n")
```

##### ➖ Difference (`-`)

- The `difference()` method:
  - Keeps items from the **first set that are NOT in the other set(s)**
  - Returns a **new set**

- **Set Operator:** `-`
  - Same result as `difference()`

```python
#Difference - The difference() method will return a new set that will contain only the items from the first set that are not present in the other set.

set1 = {"apple", 1,  "banana", 0, "cherry"}
set2 = {False, "google", 1, "apple", 2, True}

print("SET-1: ", set1)
print("SET-2: ", set2)
print("--------")

set3 = set1.difference(set2)
print("set1.difference(set2): Return a new set with elements in the set that are not in the others - ", set3)
print("--------")

set5 = set1 - set2
print("operatr 'set1 `-` set2': Return a new set with elements in the set that are not in the others - ",set5)
print("--------")
print("--------")

set6 = set2.difference(set1)
print("set2.difference(set1): Return a new set with elements in the set that are not in the others - ", set6)
print("--------")

set4 = set2 - set1
print("operatr 'set2 `-` set1': Return a new set with elements in the set that are not in the others - ",set4)
```

##### ➖ `difference_update()`

- Keeps items from the **first set that are NOT in the other set(s)**
- **Modifies the original set**
- Returns `None`
- Equivalent to the `-=` operator

🧠 Meaning:
> “Remove all common elements from the first set.”

```python
#The difference_update() method will also keep the items from the first set that are not in the other set, but it will change the original set instead of returning a new set.
set1 = {"apple", 1,  "banana", 0, "cherry"}
set2 = {False, "google", 1, "apple", 2, True}

print("SET-1: ", set1)
print("SET-2: ", set2)
print("--------")

set1.difference_update(set2) 
print("set1.difference_update(set2): Update the set, removing elements found in others - ", set1)
print("--------")

set1 = {"apple", 1,  "banana", 0, "cherry"}
set2 = {False, "google", 1, "apple", 2, True}

set1 -= set2
print("set1 -= set2: Update the set, removing elements found in others - ", set1)
print("--------")
print("--------")

set1 = {"apple", 1,  "banana", 0, "cherry"}
set2 = {False, "google", 1, "apple", 2, True}

set2.difference_update(set1) 
print("set2.difference_update(se1): Update the set, removing elements found in others - ", set2)
print("--------")

set1 = {"apple", 1,  "banana", 0, "cherry"}
set2 = {False, "google", 1, "apple", 2, True}

set2 -= set1
print("set2 -= set1: Update the set, removing elements found in others - ", set2)
```

##### 🔀 Symmetric Difference (`^`)

- The `symmetric_difference()` method:
  - Keeps **all items EXCEPT the duplicates**
  - Returns a **new set**

- **Set Operator:** `^`
  - Works **only with sets**
  - Same result as `symmetric_difference()`

```python
#The symmetric_difference() method will keep only the elements that are NOT present in both sets.
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
print("SET-1: ", set1)
print("SET-2: ", set2)
print("--------")
set3 = set1.symmetric_difference(set2)
print("set1.symmetric_difference(set2): Return a new set with elements in either the set or other but not both - ", set3)
print("--------")
set4 = set2.symmetric_difference(set1)
print("set2.symmetric_difference(set2): Return a new set with elements in either the set or other but not both - ", set3)
print("--------")
set5 = set1 ^ set2
print("set1 ^ set2:  Return a new set with elements in either the set or other but not both - ", set5)
print("--------")
```

##### 🔀 `symmetric_difference_update()`

- Keeps **all items EXCEPT the common ones**
- **Modifies the original set**
- Returns `None`
- Equivalent to the `^=` operator

🧠 Meaning:
> “Remove the intersection, keep only unique items.”

```python
#The symmetric_difference_update() method will also keep all but the duplicates, but it will change the original set instead of returning a new set.
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
print("SET-1: ", set1)
print("SET-2: ", set2)
print("--------")
set1.symmetric_difference_update(set2)
print("set1.symmetric_difference_update(set2): Update the set, keeping only elements found in either set, but not in both - ", set1)
print("--------")
set2.symmetric_difference_update(set1)
print("set2.symmetric_difference_update(set1): Update the set, keeping only elements found in either set, but not in both - ", set2)

print("--------")
set2 ^= set1
print("set2 ^= set1: Update the set, keeping only elements found in either set, but not in both - ", set2)
print("--------")
set1 ^= set2
print("set1 ^= set2: Update the set, keeping only elements found in either set, but not in both - ", set1)
```

### Frozenset

A **frozenset** is an immutable version of a set.

- Like sets, it contains **unique**, **unordered**, and **unchangeable** elements
- Unlike sets, elements **cannot be added or removed**
- All operations on a frozenset return a **new frozenset**
- Supports **non-mutating set operations** only


#### 🧠 Memory Rules (Very Important)
- **Frozenset = Immutable Set**
- ❌ No `add()`, `remove()`, `update()`
- ✅ Only operations that **return new frozensets**
- Supports **set operators**, but never `_update` methods

---

#### 🔑 One-Line Cheat Code

> **Set can change, Frozenset can’t — only new ones are created.**

```python
# Use the frozenset() constructor to create a frozenset from any iterable.

x = frozenset({"apple", "banana", "cherry"})
print("frozenset({}): Build an immutable unordered collection of unique elements - ", x)
print("frozenset({}): type is - ",type(x))

```
