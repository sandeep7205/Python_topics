## Dictionary { key : value }

Dictionaries are used to store data values in **key : value** pairs.

A dictionary is a collection that is:

- **Ordered**
- **Changeable (mutable)**
- **Does not allow duplicate keys**
- **Allows duplicate values**

---

### 📌 Ordered or Unordered?

- As of **Python 3.7+**, dictionaries are **ordered**
- In **Python 3.6 and earlier**, dictionaries are **unordered**

**Ordered** means:
- Items have a **defined insertion order**
- The order **will not change** once items are added

**Unordered** means:
- Items do **not** have a defined order
- Items **cannot be accessed by index**
- Items are accessed using **keys**

---

### 🔁 Changeable

Dictionaries are **changeable**, which means:

- You can **change** existing values
- You can **add** new key–value pairs
- You can **remove** key–value pairs

All this can be done **after the dictionary is created**.

---

### 🚫 Duplicates Not Allowed

- Dictionaries **cannot have duplicate keys**
- If the same key is used again, the **latest value overwrites the old one**
- **Values can be duplicated**, but **keys must be unique**

---

### 🧠 One-Line Memory Tip

> **Dictionary = Ordered key–value mapping with fast lookup**

#### Access Dictionary Items
- Access the items of a dictionary by referring to its key name, inside square brackets:
- There is also a method called `get()` that will give `items of a dictionary by referring to its key name`.
- The `keys()` method will return a `list of all the keys` in the dictionary.
- The `values()` method will return a `list of all the values` in the dictionary.
- The `items()` method will return `each item in a dictionary, as tuples in a list`.
- if a `specified key is present` in a dictionary use the `in` keyword:


```python
car = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  "colors": ["red", "white", "blue"],
  "electric": False,
}
print("dictionary{}  - ", car)
print("--------")

x = car["model"]
print("car['model']: Access the value of the car model - ", x)
print("--------")

##Get the value of the "electric" key using get()
yx = car.get("electric") 
print("car.get('electric'): Access the value of the car electric - ", yx) 
print("--------")

# # Get Keys
x_key = car.keys() #Get a list of the keys:
print("car.keys():list of all the keys in the dictionary - ", x_key) 
print("--------")
#  # Add a new item to the original dictionary
car["fuel_type"] = "Petrol"
print("car['fuel_type'] = 'Petrol': after new item - ", car) 
print("--------")

#Get Values
x_value = car.values() #Get a list of the values:
print("car.values(): list of all the values in the dictionary - ", x_value) 
print("--------")

##change the value of an key
car["colors"] = "red"
print("car['colors'] = 'red': change the value of an key - ", car)
print("--------")

# #Add a new item to the original dictionary
car["series"] = "GT500" 
print("car['series'] = 'GT500': Add a new item to the original dictionary - ", car)
print("--------")

#Get Items
x_items = car.items()
print("car.items(): each item in a dictionary, as tuples in a list - ", x_items)
print("--------")

#Check if Key Exists
if "model" in car:
  print("Yes, 'model' is one of the keys in the thisdict dictionary")
```

#### Add/Change Dictionary Items
- The `update()`
    - update the dictionary with the items from the given argument.
    - If the item does not exist, the item will be added.
    - The argument must be a dictionary, or an iterable object with key:value pairs.





```python
car = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  "colors": ["red", "white", "blue"],
  "electric": False,
}
print("dictionary{}  - ", car)
print("--------")

#Add item to the dictionary by using the update() method with key:value pairs.
car.update({"fuel_type": "Petrol"}) 
print("update(): Add a new item into dictionary - ", car)
print("--------")

car.update({"year": 2020}) #Update value by using the update() method with key:value pairs.
print("update(): Update the value of existing item using it's key in dictionary - ", car)
```

#### Remove Dictionary Items
- The `pop()` method `removes` the `item with the specified key name`.
- The `popitem()` method `removes the last inserted item` (in versions before 3.7, a random item is removed instead):
- The `del` keyword r`emoves the item with the specified key name` or `can also delete the dictionary` completely:
- The `clear()` method `empties the dictionary`

```python
car = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  "colors": ["red", "white", "blue"],
  "electric": False,
}
print("dictionary{}  - ", car)
print("--------")

# #The pop() method removes the item with the specified key name:
car.pop("model") 
print(".pop(): removes the item with the specified key name - ", car)
print("--------")

# #The popitem() method removes the last inserted item
car.popitem()
print(".popitem(): removes the last item - ", car)
print("--------")

## removes the item with the specified key name:
del thisdict["model"]
print("del thisdict['year']: removes the item with the specified key name - ", thisdict)
print("--------")

##The clear() method empties the dictionary:
car.clear() 
print(".clear(): empties the dictionary - ",car)
print("--------")

# #The del keyword removes the item with the specified key name:
# del car
# print(car)
```

#### Loop Dictionaries
- You can loop through a dictionary by using a `for` loop.
    - When looping through a dictionary, the return value are the keys of the dictionary, but there are methods to return the values as well.

- `values()` method to `return values` of a dictionary:
- `keys()` method to `return the keys` of a dictionary:
- `items()` method to `return both keys and values`:

```python
car = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  "colors": ["red", "white", "blue"],
  "electric": False,
}
print("dictionary{}  - ", car)
print("--------")
## Print all key names in the dictionary, one by one:
print("x: All key names in the dictionary:")
for x in car:
  print("\t-",x) 
print("--------")

## Print all values in the dictionary, one by one:
print("car[x]: All key's values in the dictionary:")
for x in car:
  print("\t-",car[x]) 
print("--------")
```

```python
#the values() method to return values of a dictionary:
car = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  "colors": ["red", "white", "blue"],
  "electric": False,
}
print("dictionary{}  - ", car)
print("--------")

##return the values of a dictionar
print("car.values(): All values from the dictionary:")

for x in car.values(): 
    print("\t-", x)

print("--------")

##return the keys of a dictionar
print("car.keys(): All keys from the dictionary:")
for x in car.keys(): 
  print("\t-", x)

print("--------")

##return both keys and values:
print("car.items(): Both keys and values from the dictionary:")
for x, y in car.items(): 
  print("\t-", x, ":", y)
```

#### Nested Dictionaries

> A dictionary can contain dictionaries, this is called nested dictionaries.

```python
myfamily = {
  "child1" : {
    "name" : "Emil",
    "year" : 2004
  },
  "child2" : {
    "name" : "Tobias",
    "year" : 2007
  },
  "child3" : {
    "name" : "Linus",
    "year" : 2011
  }
}

print("dictionary that contain multiple dictionaries - ", myfamily)
```

```python
## Create three dictionaries, then create one dictionary that will contain the other three dictionaries:

child1 = {
  "name" : "Emil",
  "year" : 2004
}
child2 = {
  "name" : "Tobias",
  "year" : 2007
}
child3 = {
  "name" : "Linus",
  "year" : 2011
}

myfamily = {
  "child1" : child1,
  "child2" : child2,
  "child3" : child3
}

print(myfamily)

print("------------------")

print(" myfamily['child2']['name'] :Access Items in Nested Dictionaries the name of child 2:", myfamily["child2"]["name"]) #Print the name of child 2
```

#### Loop Through Nested Dictionaries

- loop through a dictionary by using the `items()` method like this:


```python
for x, obj in myfamily.items():
  print(x)
  print(obj)
  print("------------------")
  for y in obj:
    print(y + ':', obj[y])
  print("------------------")
```

#### Dictionary – Methods

| Method | Description | Modifies Original |
|------|------------|------------------|
| `clear()` | Removes all elements from the dictionary | ✅ Yes |
| `copy()` | Returns a shallow copy of the dictionary | ❌ No |
| `fromkeys(keys, value)` | Creates a new dictionary from given keys and value | ❌ No |
| `get(key)` | Returns value of the key (no error if key missing) | ❌ No |
| `items()` | Returns a view of key–value pairs | ❌ No |
| `keys()` | Returns a view of all keys | ❌ No |
| `values()` | Returns a view of all values | ❌ No |
| `pop(key)` | Removes element with the specified key | ✅ Yes |
| `popitem()` | Removes the last inserted key–value pair | ✅ Yes |
| `setdefault(key, default)` | Returns value; inserts key if missing | ✅ Yes |
| `update(dict)` | Updates dictionary with key–value pairs | ✅ Yes |
