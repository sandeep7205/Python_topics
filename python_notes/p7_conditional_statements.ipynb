# Python If ... Else

### Python Conditions and `if` Statements

Python supports the usual logical conditions from mathematics:

- **Equals** → `a == b`
- **Not Equals** → `a `!=` b`
- **Less than** → `a < b`
- **Less than or equal to** → `a `<=` b`
- **Greater than** → `a > b`
- **Greater than or equal to** → `a `>=` b`

---

- The `if` Statement
    - Written using the `if` keyword
    - Evaluates a condition (`True` or `False`)
    - If **True** → code block runs  
    - If **False** → code block is skipped
    - Uses **colon (`:`)** and **indentation** (not `{ }`) to define blocks
      - Python relies on indentation (whitespace) instead of curly brackets
    - `==` is for **comparison**, `=` is for **assignment**

> **Note:** You can use spaces or tabs for indentation, but the indentation must be **consistent** within the same code block.

---

### 🔢 Truth Values in Python (`Truthy` & `Falsy`)

Python can evaluate many values as `True` or `False` in an `if` statement.

**Treated as `False`:**
- `0`
- `""` (empty string)
- `None`
- Empty collections (`[]`, `{}`, `()`)

**Treated as `True`:**
- Positive numbers (`5`)
- Negative numbers (`-3`)
- Any non-empty string (even `"False"`)

---

### 🔍 Mental Picture

Think of an `if` statement like:

> **A decision gate 🚦**  
> If the condition is true → go ahead  
> If false → skip



```python
a = 33
b = 200
if b > a:
  print(f"b = {b} is greater[>] than a = {a}")

print("--------")

## Multiple statements in an if block:
age = 20
limit = 18
if age >= limit:
  print(f"Your age is {age}, which is >= {limit}.")
  print("You are an adult")
  print("You can vote")
  print("You have full legal rights")

print("--------")

## Using a boolean variable:
is_logged_in = True
if is_logged_in:
  print(f"Login: {is_logged_in} - Welcome back!")
```

### `elif` Statement

- `elif` means **“else if”**
- Used when the previous `if` condition is **False**
- Allows checking **multiple conditions**
- Python checks conditions **top to bottom**
- As soon as a condition is **True**, its block runs
- Remaining conditions are **skipped**

> **Important:** Only the **first true** condition is executed.

🧠 **Use `elif`** when conditions are **mutually exclusive** (only one can be true).


```python
a = 33
b = 33
if b > a:
  print(f"b = {b} is greater[>] than a = {a}")
elif a == b:
  print(f"a = {a} and b = {b} are equal[==]")

print("--------")

## age groups:

age = 25
print(f"Your age is {age}.")
if age < 13:
  print("You are a child")
elif age < 20:
  print("You are a teenager")
elif age < 65:
  print("You are an adult")
elif age >= 65:
  print("You are a senior")
```

### `else` Statement

- `else` catches anything **not handled** by `if` or `elif`
- Executes when **all conditions are False**
- Acts as a **default / fallback** block
- Can be used **with or without `elif`**
- Must always come **last**
- No `elif` is allowed after `else`

🧠 **Use `else`** for default actions, validation, or error handling.

```python
a = 200
b = 33

if b > a:
  print(f"b = {b} is greater[>] than a = {a}")
elif a == b:
  print(f"a = {a} and b = {b} are equal[==]")
else:
  print(f"a = {a} is greater[>] than b = {b}")

print('--------')

## You can also have an else without the elif:
username = "Emil"
if len(username) > 0:
  print(f"Welcome, {username}!")
else:
  print("Error: Username cannot be empty")

print('--------')

## Temperature classifier:

temperature = 22
print("Temperature classifier:", temperature)
if temperature > 30:
  print("It's hot outside!")
elif temperature > 20:
  print("It's warm outside")
elif temperature > 10:
  print("It's cool outside")
else:
  print("It's cold outside!")
```


### Short Hand `if`

- Used when there is **only one statement**
- Written on the **same line**
- Colon (`:`) is still required

---

### Short Hand `if ... else` (Ternary Operator)

- Used when there is **one statement for `if` and one for `else`**
- Also called **Conditional Expression**
- Often used for **quick value assignment**

**Syntax:**
`variable = value_if_true if condition else value_if_false`

---

### When to Use Short Hand `if`

- Condition is **simple**
- Improves **readability**
- Useful for **quick decisions**
- Ternary operators are particularly useful for simple assignments and return statements.

⚠️ Avoid shorthand `if` for complex logic.  
Use normal `if–elif–else` for multi-line or complex code.

---

### 🔍 Mental Picture

Think of conditional statements like:

> **A decision gate 🚦**  
> True → execute  
> False → skip

```python
## One-line if statement:
a = 200
b = 33
if a > b: print(f"a = {a} is greater[>] than b = {b}")

print('--------')

## One-line if/else that prints a value:
a = 2
b = 330
print("A =", a, "is big from B =", b) if a > b else print("B =", b, "is big from A =",a)

print('--------')

## one-line if/else to choose a value and assign it to a variable
bigger = a if a > b else b
print("Bigger is", bigger)

print('--------')

## One line, three outcomes:
a = 30
b = 330
print("A") if a > b else print("=") if a == b else print("B")

print('--------')

## Ternary operators are particularly useful for simple assignments and return statements.
a = 200
b = 33
c = 500

if a > b and c > a:
  print("Both conditions are True")

if a > b or a > c:
  print("At least one of the conditions is True")

if not a < b:
  print("b is NOT greater than a")

if b > a:
  pass

print('--------')
## Setting a default value:

username = ""
display_name = username if username else "Guest"
print("Welcome,", display_name)
```
