# Python If ... Else

### Python Conditions and `if` Statements

Python supports the usual logical conditions from mathematics:

- **Equals** → `a == b`
- **Not Equals** → `a `!=` b`
- **Less than** → `a < b`
- **Less than or equal to** → `a `<=` b`
- **Greater than** → `a > b`
- **Greater than or equal to** → `a `>=` b`

---

- The `if` Statement
    - Written using the `if` keyword
    - Evaluates a condition (`True` or `False`)
    - If **True** → code block runs  
    - If **False** → code block is skipped
    - Uses **colon (`:`)** and **indentation** (not `{ }`) to define blocks
      - Python relies on indentation (whitespace) instead of curly brackets
    - `==` is for **comparison**, `=` is for **assignment**

> **Note:** You can use spaces or tabs for indentation, but the indentation must be **consistent** within the same code block.

---

### 🔢 Truth Values in Python (`Truthy` & `Falsy`)

Python can evaluate many values as `True` or `False` in an `if` statement.

**Treated as `False`:**
- `0`
- `""` (empty string)
- `None`
- Empty collections (`[]`, `{}`, `()`)

**Treated as `True`:**
- Positive numbers (`5`)
- Negative numbers (`-3`)
- Any non-empty string (even `"False"`)

---

### 🔍 Mental Picture

Think of an `if` statement like:

> **A decision gate 🚦**  
> If the condition is true → go ahead  
> If false → skip



`````python
a = 33
b = 200
if b > a:
  print(f"b = {b} is greater[>] than a = {a}")

print("--------")

## Multiple statements in an if block:
age = 20
limit = 18
if age >= limit:
  print(f"Your age is {age}, which is >= {limit}.")
  print("You are an adult")
  print("You can vote")
  print("You have full legal rights")

print("--------")

## Using a boolean variable:
is_logged_in = True
if is_logged_in:
  print(f"Login: {is_logged_in} - Welcome back!")
`````

### `elif` Statement

- `elif` means **“else if”**
- Used when the previous `if` condition is **False**
- Allows checking **multiple conditions**
- Python checks conditions **top to bottom**
- As soon as a condition is **True**, its block runs
- Remaining conditions are **skipped**

> **Important:** Only the **first true** condition is executed.

🧠 **Use `elif`** when conditions are **mutually exclusive** (only one can be true).


`````python
a = 33
b = 33
if b > a:
  print(f"b = {b} is greater[>] than a = {a}")
elif a == b:
  print(f"a = {a} and b = {b} are equal[==]")

print("--------")

## age groups:

age = 25
print(f"Your age is {age}.")
if age < 13:
  print("You are a child")
elif age < 20:
  print("You are a teenager")
elif age < 65:
  print("You are an adult")
elif age >= 65:
  print("You are a senior")
`````

### `else` Statement

- `else` catches anything **not handled** by `if` or `elif`
- Executes when **all conditions are False**
- Acts as a **default / fallback** block
- Can be used **with or without `elif`**
- Must always come **last**
- No `elif` is allowed after `else`

🧠 **Use `else`** for default actions, validation, or error handling.

`````python
a = 200
b = 33

if b > a:
  print(f"b = {b} is greater[>] than a = {a}")
elif a == b:
  print(f"a = {a} and b = {b} are equal[==]")
else:
  print(f"a = {a} is greater[>] than b = {b}")

print('--------')

## You can also have an else without the elif:
username = "Emil"
if len(username) > 0:
  print(f"Welcome, {username}!")
else:
  print("Error: Username cannot be empty")

print('--------')

## Temperature classifier:

temperature = 22
print("Temperature classifier:", temperature)
if temperature > 30:
  print("It's hot outside!")
elif temperature > 20:
  print("It's warm outside")
elif temperature > 10:
  print("It's cool outside")
else:
  print("It's cold outside!")
`````

#### Short Hand `if`

- Used when there is **only one statement**
- Written on the **same line**
- Colon (`:`) is still required

---

#### Short Hand `if ... else` (Ternary Operator)

- Used when there is **one statement for `if` and one for `else`**
- Also called **Conditional Expression**
- Often used for **quick value assignment**

**Syntax:**
`variable = value_if_true if condition else value_if_false`

---

> When to Use Short Hand `if`

- Condition is **simple**
- Improves **readability**
- Useful for **quick decisions**
- Ternary operators are particularly useful for simple assignments and return statements.

⚠️ Avoid shorthand `if` for complex logic.  
Use normal `if–elif–else` for multi-line or complex code.

---

> 🔍 Mental Picture

Think of conditional statements like:

> **A decision gate 🚦**  
> True → execute  
> False → skip

`````python
## One-line if statement:
a = 200
b = 33
if a > b: print(f"a = {a} is greater[>] than b = {b}")

print('--------')

## One-line if/else that prints a value:
a = 2
b = 330
print("A =", a, "is big from B =", b) if a > b else print("B =", b, "is big from A =",a)

print('--------')

## one-line if/else to choose a value and assign it to a variable
bigger = a if a > b else b
print("Bigger is", bigger)

print('--------')

## One line, three outcomes:
a = 30
b = 330
print("A") if a > b else print("=") if a == b else print("B")

print('--------')

## Ternary operators are particularly useful for simple assignments and return statements.
a = 200
b = 33
c = 500

if a > b and c > a:
  print("Both conditions are True")

if a > b or a > c:
  print("At least one of the conditions is True")

if not a < b:
  print("b is NOT greater than a")

if b > a:
  pass

print('--------')
## Setting a default value:

username = ""
display_name = username if username else "Guest"
print("Welcome,", display_name)
`````

#### Logical Operators

> Logical operators are used to **combine conditional statements**.

Python has three logical operators:

- **`and`** → Returns `True` if **both** conditions are true
- **`or`** → Returns `True` if **at least one** condition is true
- **`not`** → Reverses the result (`True` → `False`, `False` → `True`)

---

##### 🧠 Memory Tip

> **and = all true | or = any true | not = reverse**

---

**Truth Tables**

> and Operator Truth Table

```
Condition 1	Condition 2	Result
True	        True	        True
True	        False	        False
False	        True	        False
False	        False	        False
```

> or Operator Truth Table

```
Condition 1	Condition 2	Result
True	        True	        True
True	        False	        True
False	        True	        True
False	        False	        False
```

`````python
## and Operator
a = 200
b = 33
c = 500
if a > b and c > a:
  print(f"and : Both ({a} is greater than {b}, AND {c} is greater than {a}) conditions are True")

print('--------')

## or Operator
a = 200
b = 33
c = 500
if a > b or a > c:
  print(f"or : At least one of the ({a} is greater than {b}, OR if {a} is greater than {c}) conditions is True")

print('--------')

## not Operator
a = 33
b = 200
if not a > b:
  print(f"not : Reverse the result of the conditional where {a} is NOT greater than {b}")

print('--------')

## Combining and, or, and not:
age = 25
is_student = False
has_discount_code = True

if ((age < 18 or age > 65) and not is_student) or has_discount_code:
  print("Discount applies!")
`````

#### Nested If

`````python
## Nested If Statements
print("--------Nested If Statements--------")
score = 85
attendance = 90
submitted = False

print (f"score - {score} | attendance - {attendance} | assignment submited: {submitted}")
if score >= 60:
  if attendance >= 80:
    if submitted:
      print("Pass with good standing")
    else:
      print("Pass but missing assignment")
  else:
    print("Pass but low attendance")
else:
  print("Fail")

print('--------')
`````

#### Nested If vs Logical Operators

`````python
print("This nested if:")
temperature = 25
is_sunny = True

if temperature > 20:
  if is_sunny:
    print("\tPerfect beach weather!")


print("\nCould also be written with and:")

temperature = 25
is_sunny = True

if temperature > 20 and is_sunny:
  print("\t\t\t Perfect beach weather!")
`````

#### `pass` Statement

- `if` statements cannot be empty  
- Use `pass` to avoid syntax errors when no code is needed
- `pass` is a **null operation** → nothing happens when it runs
- Acts as a **placeholder** in the code

---

> When to Use `pass`

- While creating code structure before logic is written
- When a statement is required syntactically but no action is needed
- As a placeholder for future implementation
- In empty `if` statements, loops, functions, or classes


`````python
value = 50

if value < 0:
  print("Negative value")
elif value == 0:
  pass # Zero case - no action needed
else:
  print("Positive value")
`````

#### `match` Statement

- Used to perform **different actions based on different values**
- An alternative to multiple `if–elif–else` statements
- Selects **one matching case** to execute

---

> Syntax

```python
match expression:
    case x:
        # code block
    case y:
        # code block
    case z:
        # code block
````

---

> How It Works

* The `match` expression is evaluated **once**
* The value is compared with each `case`
* If a match is found, the corresponding block is executed
* All remaining cases are skipped

---

> Default Case (`_`)

* Use the underscore `_` as the **last case** to act as a default
* `_` always matches any value
* Must be placed **last**, otherwise it will catch everything

```python
match value:
    case 1:
        print("One")
    case 2:
        print("Two")
    case _:
        print("No match found")
```

---

> Combine Values (`|`)

* Use the pipe `|` as an **OR operator** inside a `case`
* Allows matching **multiple values** in one case

```python
match value:
    case 1 | 3 | 5:
        print("Odd number")
    case 2 | 4 | 6:
        print("Even number")
```

---

> If Statements as Guards

* You can add an `if` condition to a `case` for extra checks
* The case runs **only if both the pattern and the guard are true**

```python
match number:
    case x if x > 0:
        print("Positive number")
    case x if x < 0:
        print("Negative number")
    case _:
        print("Zero")
```

---

> 🧠 Memory Tip

> **Pattern + Guard → Execute**
> **`_` = default | `|` = OR**

`````python
## Match Statement with Default Value
day = 7
print(f"Day is {day} and So: ")
match day:
  case 1:
    print("Monday")
  case 2:
    print("Tuesday")
  case 3:
    print("Wednesday")
  case 4:
    print("Thursday")
  case 5:
    print("Friday")
  case 6:
    print("Saturday")
  case 7:
    print("Sunday")
  case _:
    print("Looking forward to the Weekend")

print('--------')

## pipe character | as an or operator in the Match Statement for case evaluation
day = 6
print(f"Day is {day} and So: ")
match day:
  case 1 | 2 | 3 | 4 | 5:
    print("Today is a weekday")
  case 6 | 7:
    print("I love weekends!")
  case _:
    print("Looking forward to the Weeks")

print('--------')

## if statements in the Match Statement for case evaluation
month = 5
day = 4
print(f"Month is {month} and Day is {day} and So: ")
match day:
  case 1 | 2 | 3 | 4 | 5 if month == 4:
    print("A weekday in April")
  case 1 | 2 | 3 | 4 | 5 if month == 5:
    print("A weekday in May")
  case _:
    print("No match")
`````
